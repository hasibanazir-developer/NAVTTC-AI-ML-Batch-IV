What is an Exception

```python
# Topic: What is an Exception
# An exception is an error that happens while the program is running.
# If we do not handle it, the program stops and shows a red error message.
# Below we handle it so the program keeps running.

try:
    print(10 / 0)
except ZeroDivisionError as error:
    print("Something went wrong:", error)

print("The program is still running")
```

Python try...except Block

```python
# Topic: try...except Block
# Code that might cause an error goes inside try.
# If an error happens, Python jumps to except instead of crashing.

try:
    total_marks = 450
    subjects = 0
    average = total_marks / subjects   # dividing by zero causes an error
    print("Average:", average)
except:
    print("Error: number of subjects cannot be zero")
```

Example: Exception Handling Using try...except

```python
# Topic: Example of try...except
# Here we safely divide each number in a list by a divisor

divisors = [5, 0, 2]

for d in divisors:
    try:
        result = 100 / d
        print("100 divided by", d, "is", result)
    except ZeroDivisionError:
        print("Cannot divide 100 by", d)
```

Catching Specific Exceptions

```python
# Topic: Catching Specific Exceptions
# We can write separate except blocks for different types of errors.
# Python runs only the block that matches the error.

items = ["pen", "book", "bag"]

def get_item(position):
    try:
        print("Item at position", position, "is", items[position])
    except IndexError:
        print("Error: position", position, "does not exist in the list")
    except TypeError:
        print("Error: position must be a whole number, not", repr(position))

get_item(1)      # works fine
get_item(10)     # IndexError
get_item("two")  # TypeError
```

Catching Multiple Exceptions in One Block

```python
# Topic: Handling Several Errors Together
# A tuple of error types lets one except block handle all of them

values = ["25", "abc", "0"]

for v in values:
    try:
        number = int(v)
        print("100 /", number, "=", 100 / number)
    except (ValueError, ZeroDivisionError) as error:
        print("Could not use", repr(v), "->", type(error).__name__, ":", error)
```

Python try with else Clause

```python
# Topic: try...else
# The else block runs only when the try block finishes WITHOUT any error

def check_even(num):
    try:
        if num % 2 != 0:
            raise ValueError("number is odd")
    except ValueError as error:
        print(num, "-> Error:", error)
    else:
        # This runs only if no error happened above
        print(num, "-> It is an even number, so 1 /", num, "=", 1 / num)

check_even(4)
check_even(7)
```

Errors Inside else are Not Caught

```python
# Topic: Errors in the else Block
# If an error happens inside else, the except blocks above do NOT catch it.
# Here we wrap everything in an outer try just to show this safely.

num = 0

try:
    try:
        if num % 2 != 0:
            raise ValueError("number is odd")
    except ValueError:
        print("Inner except: odd number")
    else:
        print("0 is even, now dividing...")
        print(1 / num)   # this error happens inside else
except ZeroDivisionError:
    print("The ZeroDivisionError came from the else block, so only the outer try caught it")
```

Python try...finally

```python
# Topic: try...finally
# The finally block ALWAYS runs, whether an error happens or not.
# It is often used for clean-up work, like closing a file.

def divide(a, b):
    try:
        print("Result:", a / b)
    except ZeroDivisionError:
        print("Error: cannot divide by zero")
    finally:
        print("Finally block: division attempt finished\n")

divide(20, 4)   # no error
divide(20, 0)   # error
```

Using finally with a File

```python
# Topic: Practical Use of finally
# Even if something goes wrong while reading, the file still gets closed

with open("notes.txt", "w") as f:
    f.write("Week 2: Exception Handling")

file = open("notes.txt")
try:
    print("File content:", file.read())
    print(int("not a number"))   # this line causes an error on purpose
except ValueError:
    print("Error: could not convert text to a number")
finally:
    file.close()
    print("Is the file closed?", file.closed)
```
