What is a User-Defined (Custom) Exception

```python
# Topic: User-Defined Exceptions
# Python has many built-in errors, but sometimes we need our own error
# for our program's rules. These are called custom or user-defined exceptions.

# Example rule: marks must be between 0 and 100.
# Python does not know this rule, so 150 is not an error by default.
marks = 150
print("Python accepts marks =", marks, "without any error")
print("So we need our own exception to catch this kind of mistake")
```

Defining a Custom Exception

```python
# Topic: Defining a Custom Exception
# A custom exception is a class that inherits from Python's Exception class.
# 'pass' means we do not add anything extra to it.

class InvalidMarksError(Exception):
    """Raised when marks are not between 0 and 100"""
    pass

# raise is used to throw our own exception
try:
    raise InvalidMarksError("Marks must be between 0 and 100")
except InvalidMarksError as error:
    print("Caught custom error:", error)
```

Example: Python User-Defined Exception

```python
# Topic: Using a Custom Exception in a Program
# Rule: a person must be at least 18 years old to register for the course

class UnderAgeError(Exception):
    """Raised when the age is below the required age"""
    pass

REQUIRED_AGE = 18

def register(name, age):
    try:
        if age < REQUIRED_AGE:
            raise UnderAgeError
        print(name, "is registered successfully")
    except UnderAgeError:
        print(name, "cannot register: age", age, "is below", REQUIRED_AGE)

# We use fixed values instead of input() so the notebook runs on its own
register("Hasiba", 21)
register("Ahmed", 15)
```

Custom Exception with a Message

```python
# Topic: Passing a Message to a Custom Exception

class InvalidMarksError(Exception):
    pass

def save_marks(subject, marks):
    if marks < 0 or marks > 100:
        raise InvalidMarksError(f"{marks} is not valid for {subject}. Use 0 to 100.")
    print(subject, "marks saved:", marks)

for subject, marks in [("Python", 92), ("Statistics", 120), ("AI", -5)]:
    try:
        save_marks(subject, marks)
    except InvalidMarksError as error:
        print("Error:", error)
```

Customizing Exception Classes

```python
# Topic: Customizing an Exception Class
# We can add our own __init__ method to store extra information.
# super().__init__() passes the message to the parent Exception class.

class SalaryNotInRangeError(Exception):
    """Raised when the salary is outside the allowed range"""

    def __init__(self, salary, message="Salary is not in the range 30,000 to 300,000"):
        self.salary = salary       # keep the wrong value for later use
        self.message = message
        super().__init__(self.message)

    def __str__(self):
        # This controls what is shown when the error is printed
        return f"{self.salary} -> {self.message}"

def set_salary(amount):
    if not 30000 <= amount <= 300000:
        raise SalaryNotInRangeError(amount)
    print("Salary set to", amount)

for amount in [85000, 12000, 500000]:
    try:
        set_salary(amount)
    except SalaryNotInRangeError as error:
        print("Error:", error)
        print("Wrong value stored in the exception:", error.salary)
```

Hierarchy of Custom Exceptions

```python
# Topic: A Family of Custom Exceptions
# In bigger programs we make one base exception and several child exceptions.
# Catching the base class catches all of its children too.

class BankError(Exception):
    """Base class for all bank errors"""
    pass

class InsufficientBalanceError(BankError):
    pass

class InvalidAmountError(BankError):
    pass

def withdraw(balance, amount):
    if amount <= 0:
        raise InvalidAmountError("Amount must be greater than zero")
    if amount > balance:
        raise InsufficientBalanceError(f"Balance is {balance}, cannot withdraw {amount}")
    return balance - amount

for amount in [500, -100, 5000]:
    try:
        new_balance = withdraw(2000, amount)
        print("Withdrew", amount, "| New balance:", new_balance)
    except BankError as error:
        print(type(error).__name__, "->", error)
```
