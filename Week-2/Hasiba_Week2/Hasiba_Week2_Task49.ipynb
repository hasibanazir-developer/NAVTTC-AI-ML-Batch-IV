Mean, Median and Mode: Introduction

```python
# Topic: Mean, Median and Mode
# These three values describe the "center" of a set of numbers.
# We will use the daily temperatures (in Celsius) of Lahore for two weeks.

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]

print("Temperatures:", temps)
print("Number of days:", len(temps))
```

Mean (Calculated by Hand)

```python
# Topic: Mean
# Mean is the average: add all values and divide by how many values there are

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]

total = sum(temps)
count = len(temps)
mean_value = total / count

print("Sum of temperatures:", total)
print("Number of days:", count)
print("Mean temperature:", round(mean_value, 2))
```

Mean using NumPy

```python
# Topic: Mean with NumPy
# NumPy has a mean() function that does the same calculation for us

import numpy as np

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]

mean_value = np.mean(temps)
print("Mean temperature using NumPy:", round(mean_value, 2))
```

Median (Odd Number of Values)

```python
# Topic: Median
# Median is the middle value after sorting the numbers.
# With an odd count, there is exactly one middle value.

import numpy as np

prices = [120, 80, 150, 95, 200, 110, 90]   # 7 values

print("Sorted prices:", sorted(prices))
print("Median price:", np.median(prices))
```

Median (Even Number of Values)

```python
# Topic: Median with an Even Count
# With an even count there are two middle values,
# so the median is the average of those two.

import numpy as np

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]   # 14 values

sorted_temps = sorted(temps)
middle_left = sorted_temps[len(temps) // 2 - 1]
middle_right = sorted_temps[len(temps) // 2]

print("Sorted temperatures:", sorted_temps)
print("Two middle values:", middle_left, "and", middle_right)
print("Median by hand:", (middle_left + middle_right) / 2)
print("Median using NumPy:", np.median(temps))
```

Mode using SciPy

```python
# Topic: Mode
# Mode is the value that appears the most times.
# The SciPy library has a stats.mode() function for this.

from scipy import stats

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]

result = stats.mode(temps)
print("Most common temperature (mode):", result.mode)
print("It appears", result.count, "times")
```

Mode using Python's statistics Module

```python
# Topic: Mode without SciPy
# Python also has a built-in statistics module that can find the mode,
# and Counter can show how many times every value appears

import statistics
from collections import Counter

temps = [31, 33, 29, 35, 33, 30, 28, 33, 36, 31, 29, 34, 33, 30]

print("Mode using statistics:", statistics.mode(temps))
print("How often each value appears:", Counter(temps).most_common())
```

Comparing Mean, Median and Mode

```python
# Topic: Mean vs Median with an Outlier
# An outlier is a value that is very different from the others.
# The mean changes a lot because of it, but the median stays almost the same.

import numpy as np
from scipy import stats

salaries = [40000, 42000, 45000, 45000, 48000, 50000]
with_outlier = salaries + [500000]   # one very high salary

for label, data in [("Normal salaries", salaries), ("With one outlier", with_outlier)]:
    print(label)
    print("  Mean  :", round(np.mean(data), 2))
    print("  Median:", np.median(data))
    print("  Mode  :", stats.mode(data).mode)
```
