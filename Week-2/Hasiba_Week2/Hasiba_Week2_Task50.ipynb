What is Standard Deviation

```python
# Topic: Standard Deviation
# Standard deviation tells us how spread out the values are.
# Low standard deviation: values are close to the mean.
# High standard deviation: values are spread far from the mean.

import numpy as np

# Daily study hours of two students over one week
student_a = [4, 5, 4, 5, 4, 5, 4]     # very regular
student_b = [1, 8, 2, 9, 0, 7, 3]     # very irregular

print("Student A -> mean:", round(np.mean(student_a), 2), "| std:", round(np.std(student_a), 2))
print("Student B -> mean:", round(np.mean(student_b), 2), "| std:", round(np.std(student_b), 2))
print("Student B has a much higher standard deviation, so the hours are more spread out")
```

Variance (Calculated Step by Step)

```python
# Topic: Variance
# Variance is another way to measure spread. Steps:
# 1. Find the mean
# 2. Subtract the mean from each value
# 3. Square each difference
# 4. Find the average of the squared differences

scores = [60, 72, 85, 90, 68]

# Step 1: mean
mean = sum(scores) / len(scores)
print("Step 1 - Mean:", mean)

# Step 2: differences from the mean
diffs = [s - mean for s in scores]
print("Step 2 - Differences:", diffs)

# Step 3: squared differences
squares = [round(d ** 2, 2) for d in diffs]
print("Step 3 - Squared differences:", squares)

# Step 4: average of the squares
variance = sum(squares) / len(squares)
print("Step 4 - Variance:", round(variance, 2))
```

Variance using NumPy

```python
# Topic: Variance with NumPy
# NumPy's var() function calculates variance in one line

import numpy as np

scores = [60, 72, 85, 90, 68]
print("Variance using NumPy:", round(np.var(scores), 2))
```

Standard Deviation from Variance

```python
# Topic: Standard Deviation Formula
# Standard deviation is simply the square root of the variance

import math
import numpy as np

scores = [60, 72, 85, 90, 68]

variance = np.var(scores)
std_by_hand = math.sqrt(variance)

print("Variance:", round(variance, 2))
print("Square root of variance:", round(std_by_hand, 2))
print("Standard deviation using NumPy std():", round(np.std(scores), 2))
```

Symbols

```python
# Topic: Symbols Used
# Standard deviation is written with the Greek letter sigma: σ
# Variance is written as sigma squared: σ²

import numpy as np

scores = [60, 72, 85, 90, 68]
sigma = np.std(scores)

print("σ  (standard deviation) =", round(sigma, 2))
print("σ² (variance)           =", round(sigma ** 2, 2))
```

Population vs Sample Standard Deviation

```python
# Topic: Population vs Sample
# NumPy divides by n by default (population).
# When our data is only a sample of a bigger group, we divide by n-1 using ddof=1.

import numpy as np

scores = [60, 72, 85, 90, 68]

print("Population std (ddof=0):", round(np.std(scores), 2))
print("Sample std     (ddof=1):", round(np.std(scores, ddof=1), 2))
```

Practice: Comparing Spread of Two Datasets

```python
# Topic: Practice - Which Shop Has More Stable Sales?
# Same average, but different spread

import numpy as np

shop_1 = [100, 102, 98, 101, 99, 100]
shop_2 = [60, 140, 80, 120, 70, 130]

for name, sales in [("Shop 1", shop_1), ("Shop 2", shop_2)]:
    print(name)
    print("  Mean    :", round(np.mean(sales), 2))
    print("  Variance:", round(np.var(sales), 2))
    print("  Std dev :", round(np.std(sales), 2))

print("Shop 1 has a lower standard deviation, so its sales are more stable")
```
