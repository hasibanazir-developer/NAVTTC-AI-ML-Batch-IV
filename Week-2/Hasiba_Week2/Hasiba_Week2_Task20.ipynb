Defining a Function

```python
# Topic: Defining a Function
# A function is a reusable block of code. We create it using the def keyword.

def welcome():
    # This code runs every time the function is called
    print("Welcome to the AI course!")
```

Calling a Function

```python
# Topic: Calling a Function
# To run a function, we write its name followed by brackets

def welcome():
    print("Welcome to the AI course!")

welcome()   # first call
welcome()   # we can call it as many times as we want

# A function can also be stored in a variable and called through it
say_hi = welcome
say_hi()
```

Function Arguments

```python
# Topic: Function Arguments
# Arguments are values we send into a function when we call it

def check_number(num):
    # % gives the remainder; if the remainder is 0 the number is even
    if num % 2 == 0:
        print(num, "is even")
    else:
        print(num, "is odd")

check_number(12)
check_number(7)
```

Default Arguments

```python
# Topic: Default Arguments
# A default argument is used when no value is given for it

def order_tea(cups=1):
    print("You ordered", cups, "cup(s) of tea")

order_tea()    # uses the default value 1
order_tea(3)   # uses the value we passed
```

Keyword Arguments

```python
# Topic: Keyword Arguments
# With keyword arguments, we write the name of the parameter while calling,
# so the order does not matter

def student_info(name, course):
    print(name, "is studying", course)

student_info(course="Machine Learning", name="Hasiba")
```

Positional Arguments

```python
# Topic: Positional Arguments
# Positional arguments are matched by their order, so order matters

def introduce(name, city):
    print("Hi, I am", name, "from", city)

introduce("Sara", "Lahore")   # correct order
introduce("Lahore", "Sara")   # wrong order gives a strange result
```

Arbitrary Arguments (*args and **kwargs)

```python
# Topic: Arbitrary Arguments
# *args lets a function accept any number of normal values (stored as a tuple)
# **kwargs lets a function accept any number of named values (stored as a dictionary)

def add_all(*args):
    print("Numbers received:", args)
    return sum(args)

print("Total:", add_all(2, 4, 6, 8))

def show_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

show_profile(name="Hasiba", city="Lahore", course="AI")
```

Function within a Function (Nested Function)

```python
# Topic: Nested Function
# A function can be written inside another function.
# The inner function can use variables of the outer function.

def outer_function():
    message = "Hello from the outer function"

    def inner_function():
        # The inner function can read 'message' from the outer function
        print(message)

    inner_function()   # call the inner function

outer_function()
```

Return Statement

```python
# Topic: Return Statement
# return sends a result back to the place where the function was called

def rectangle_area(length, width):
    return length * width

area = rectangle_area(5, 3)
print("Area of rectangle:", area)

# If a function has no return statement, it gives back None
def no_return():
    print("This function returns nothing")

result = no_return()
print("Returned value:", result)
```

Pass by Reference and Pass by Value

```python
# Topic: Pass by Reference vs Pass by Value
# Python passes a reference to the object.
# Mutable objects (like lists) can be changed inside a function.
# Immutable objects (like numbers) cannot be changed inside a function.

def add_item(my_list):
    my_list.append(100)   # this changes the original list

numbers = [10, 20, 30]
add_item(numbers)
print("List after function call:", numbers)

def change_number(x):
    x = x + 50            # this creates a new value only inside the function
    print("Inside function:", x)

value = 10
change_number(value)
print("Outside function:", value)
```
