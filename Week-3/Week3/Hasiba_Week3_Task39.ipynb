Sorting Arrays

```python
# Topic: np.sort()
# np.sort() returns a sorted COPY; the original array stays the same

import numpy as np

ages = np.array([34, 19, 27, 45, 22])
sorted_ages = np.sort(ages)

print("Sorted:", sorted_ages)
print("Original is unchanged:", ages)
```

Sorting in Place with the sort() Method

```python
# Topic: arr.sort()
# The sort() method changes the array itself

import numpy as np

ages = np.array([34, 19, 27, 45, 22])
ages.sort()
print("Array after sort():", ages)

# For descending order, sort and then reverse
print("Descending:", np.sort(ages)[::-1])
```

Sorting Strings

```python
# Topic: Sorting Strings
# Strings are sorted in alphabetical order

import numpy as np

cities = np.array(["Quetta", "Lahore", "Karachi", "Multan"])
print(np.sort(cities))
```

Sorting Booleans

```python
# Topic: Sorting Booleans
# False comes before True

import numpy as np

passed = np.array([True, False, True, False, True])
print(np.sort(passed))
```

Sorting a 2-D Array

```python
# Topic: Sorting a 2-D Array
# Each row (inner array) is sorted on its own

import numpy as np

grid = np.array([[9, 2, 7],
                 [4, 8, 1]])

print("Each row sorted:")
print(np.sort(grid))

print("Each column sorted (axis=0):")
print(np.sort(grid, axis=0))
```
