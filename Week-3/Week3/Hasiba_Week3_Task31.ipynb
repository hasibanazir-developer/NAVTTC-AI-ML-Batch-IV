Data Types in Python

```python
# Topic: Python's Basic Data Types
# Python has strings, integers, floats, booleans and complex numbers

values = ["AI", 25, 3.75, True, 2 + 3j]

for v in values:
    print(repr(v), "->", type(v).__name__)
```

Data Types in NumPy

```python
# Topic: NumPy Data Types
# NumPy has more data types, and each one has a short character code:
# i - integer        b - boolean        u - unsigned integer
# f - float          c - complex float  m - timedelta
# M - datetime       O - object         S - string (bytes)
# U - unicode string V - fixed chunk of memory (void)

import numpy as np

examples = {
    "i": np.array([1, 2]),
    "f": np.array([1.5, 2.5]),
    "b": np.array([True, False]),
    "U": np.array(["cat", "dog"]),
    "c": np.array([1 + 2j]),
}

for code, arr in examples.items():
    print(code, "->", arr.dtype, "| kind:", arr.dtype.kind)
```

Checking the Data Type of an Array

```python
# Topic: dtype Property
# dtype tells us the type of data stored in the array

import numpy as np

ages = np.array([18, 21, 25])
names = np.array(["Ali", "Sara", "Hasiba"])

print("ages dtype:", ages.dtype)
print("names dtype:", names.dtype)   # <U6 means unicode strings up to 6 characters
```

Creating Arrays With a Defined Data Type

```python
# Topic: Choosing the Data Type While Creating
# The dtype argument sets the type. We can also give a size, like i4 (4-byte integer).

import numpy as np

as_strings = np.array([1, 2, 3], dtype="S")
print(as_strings, "->", as_strings.dtype)

small_ints = np.array([1, 2, 3], dtype="i4")
print(small_ints, "->", small_ints.dtype)

floats = np.array([1, 2, 3], dtype=float)
print(floats, "->", floats.dtype)
```

What if a Value Can Not Be Converted

```python
# Topic: Conversion Error
# If a value cannot be changed into the chosen type, NumPy raises a ValueError

import numpy as np

try:
    arr = np.array(["10", "twenty", "30"], dtype="i")
except ValueError as error:
    print("ValueError:", error)
```

Converting Data Type on Existing Arrays

```python
# Topic: astype()
# astype() makes a NEW array with a different data type

import numpy as np

prices = np.array([99.9, 150.5, 20.2])

as_int = prices.astype("i")      # decimal part is removed
print(as_int, "->", as_int.dtype)

as_int2 = prices.astype(int)     # same thing using the Python type name
print(as_int2, "->", as_int2.dtype)

flags = np.array([0, 3, 0, 7])
as_bool = flags.astype(bool)     # 0 becomes False, anything else becomes True
print(as_bool, "->", as_bool.dtype)
```
