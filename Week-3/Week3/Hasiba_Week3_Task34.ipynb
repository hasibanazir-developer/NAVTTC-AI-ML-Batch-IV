Reshaping Arrays

```python
# Topic: Reshaping
# Reshaping means changing the shape of an array:
# adding or removing dimensions, or changing how many values are in each one

import numpy as np

months = np.arange(1, 13)   # numbers 1 to 12
print("Original:", months, "| shape:", months.shape)
```

Reshape From 1-D to 2-D

```python
# Topic: 1-D to 2-D
# 12 values can become 4 rows with 3 values each (4 x 3 = 12)

import numpy as np

months = np.arange(1, 13)
quarters = months.reshape(4, 3)   # 4 quarters, 3 months each

print(quarters)
print("Shape:", quarters.shape)
```

Reshape From 1-D to 3-D

```python
# Topic: 1-D to 3-D
# 12 values can become 2 blocks, each with 3 rows of 2 values (2 x 3 x 2 = 12)

import numpy as np

months = np.arange(1, 13)
blocks = months.reshape(2, 3, 2)

print(blocks)
print("Shape:", blocks.shape)
```

Can We Reshape Into any Shape

```python
# Topic: Reshape Rules
# The new shape must hold exactly the same number of values.
# 12 values cannot fit into a 5 x 3 shape (15 places).

import numpy as np

months = np.arange(1, 13)

try:
    months.reshape(5, 3)
except ValueError as error:
    print("Error:", error)

print("But 6 x 2 works:")
print(months.reshape(6, 2))
```

Returns Copy or View

```python
# Topic: Reshape Gives a View
# The reshaped array shares data with the original.
# If base is not None, it is a view.

import numpy as np

nums = np.array([1, 2, 3, 4, 5, 6])
grid = nums.reshape(2, 3)

print("Base of reshaped array:", grid.base)

grid[0, 0] = 100
print("After changing the reshaped array, original is:", nums)
```

Unknown Dimension

```python
# Topic: Using -1 in Reshape
# We can write -1 for ONE dimension and NumPy will work out its size

import numpy as np

nums = np.arange(1, 13)

a = nums.reshape(3, -1)       # NumPy finds 4 columns
print("reshape(3, -1) gives shape", a.shape)
print(a)

b = nums.reshape(2, 2, -1)    # NumPy finds 3
print("reshape(2, 2, -1) gives shape", b.shape)

try:
    nums.reshape(-1, -1)      # only one -1 is allowed
except ValueError as error:
    print("Error:", error)
```

Flattening the Arrays

```python
# Topic: Flattening
# Flattening turns a multi-dimensional array into a 1-D array

import numpy as np

grid = np.array([[1, 2, 3],
                 [4, 5, 6]])

print("Using reshape(-1):", grid.reshape(-1))
print("Using flatten()  :", grid.flatten())
print("Using ravel()    :", grid.ravel())
print("Using flip()     :", np.flip(grid.reshape(-1)))
```
