Import Pandas

```python
# Topic: Importing Pandas
# Once installed, we bring Pandas into our program using import

import pandas

books = {
    "Title": ["Python Basics", "Data Science 101", "AI for All"],
    "Copies": [5, 3, 8]
}

table = pandas.DataFrame(books)
print(table)
```

Pandas as pd

```python
# Topic: The pd Alias
# Most people write "import pandas as pd" so they can type pd instead of pandas

import pandas as pd

books = {
    "Title": ["Python Basics", "Data Science 101", "AI for All"],
    "Copies": [5, 3, 8]
}

table = pd.DataFrame(books)
print(table)
```

Checking Pandas Version

```python
# Topic: Checking the Version
# __version__ tells us which version of Pandas is installed

import pandas as pd

print("Installed Pandas version:", pd.__version__)
```
