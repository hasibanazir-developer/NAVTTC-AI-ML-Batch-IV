Data Cleaning

```python
# Topic: Data Cleaning
# Data cleaning means fixing bad data. Bad data can be:
# 1. Empty cells
# 2. Data in the wrong format
# 3. Wrong data
# 4. Duplicate rows
```

Our Data Set

```python
# Topic: Our Data Set
import pandas as pd
import numpy as np

# Sample data with problems put in on purpose
def make_dirty():
    return pd.DataFrame({
        "Day":      ["2026/09/01", "2026/09/02", "2026/09/03", "2026/09/04", "2026/09/05",
                     "2026/09/06", "2026/09/07", "2026/09/07", np.nan, "20260910"],
        "Hours":    [2, 3, 1.5, 40, 2.5, 3.5, 1, 1, 5, 2],
        "Sleep":    [7, 6, 8, 6.5, np.nan, 6, 8, 8, 5.5, 7.5],
        "Score":    [62, 70, np.nan, 82, 66, 76, 50, 50, 91, 60],
    })

df = make_dirty()
print(df)

# Problems in this data:
# - Empty cells: Day (row 8), Sleep (row 4), Score (row 2)
# - Wrong format: Day in row 9 is "20260910" instead of "2026/09/10"
# - Wrong data: Hours in row 3 is 40, which is impossible for one day
# - Duplicates: rows 6 and 7 are exactly the same
```

Cleaning Empty Cells: Remove Rows

```python
# Topic: Cleaning Empty Cells: Remove Rows
# dropna() removes every row that has at least one empty cell.
# It returns a NEW DataFrame and does not change the original.
df = make_dirty()
new_df = df.dropna()
print(new_df)
print("Rows before:", len(df), "| Rows after:", len(new_df))
```

Cleaning Empty Cells: inplace=True

```python
# Topic: Cleaning Empty Cells: inplace=True
# inplace=True changes the original DataFrame itself
df = make_dirty()
df.dropna(inplace=True)
print(df)
```

Cleaning Empty Cells: Replace Empty Values

```python
# Topic: Cleaning Empty Cells: Replace Empty Values
# fillna() puts a value in every empty cell instead of deleting the row
df = make_dirty()
print(df.fillna(0))
```

Cleaning Empty Cells: Replace Only for a Column

```python
# Topic: Cleaning Empty Cells: Replace Only for a Column
# We can fill empty cells in just one column
df = make_dirty()
df["Score"] = df["Score"].fillna(65)
print(df)
```

Cleaning Empty Cells: Replace Using Mean, Median, or Mode

```python
# Topic: Cleaning Empty Cells: Replace Using Mean, Median, or Mode
# A common way is to fill empty cells with the mean, median or mode of that column
df = make_dirty()

mean_score = df["Score"].mean()
median_sleep = df["Sleep"].median()
mode_score = df["Score"].mode()[0]   # mode() gives a Series, [0] takes the first value

print("Mean score:", round(mean_score, 2))
print("Median sleep:", median_sleep)
print("Mode score:", mode_score)

df["Score"] = df["Score"].fillna(mean_score)
df["Sleep"] = df["Sleep"].fillna(median_sleep)
print(df)
```

Cleaning Data of Wrong Format

```python
# Topic: Cleaning Data of Wrong Format
# All dates should be real dates. to_datetime() converts them.
# format="mixed" lets Pandas read both "2026/09/01" and "20260910".
df = make_dirty()
df["Day"] = pd.to_datetime(df["Day"], format="mixed")
print(df)
# The empty date became NaT ("Not a Time")
```

Removing Rows with Empty Dates

```python
# Topic: Removing Rows with Empty Dates
# After converting, we remove the row whose date is still empty.
# subset tells dropna() to look only at the Day column.
df = make_dirty()
df["Day"] = pd.to_datetime(df["Day"], format="mixed")
df.dropna(subset=["Day"], inplace=True)
print(df)
```

Fixing Wrong Data: Replacing Values

```python
# Topic: Fixing Wrong Data: Replacing Values
# Hours = 40 in row 3 is clearly a typing mistake (it was probably 4)
df = make_dirty()
df.loc[3, "Hours"] = 4
print(df)
```

Fixing Wrong Data: Using a Rule

```python
# Topic: Fixing Wrong Data: Using a Rule
# For big data, we use a rule: nobody studies more than 12 hours a day,
# so any value above 12 is set to 12
df = make_dirty()
for x in df.index:
    if df.loc[x, "Hours"] > 12:
        df.loc[x, "Hours"] = 12
print(df)
```

Fixing Wrong Data: Removing Rows

```python
# Topic: Fixing Wrong Data: Removing Rows
# Another way is to delete the rows that have wrong data
df = make_dirty()
for x in df.index:
    if df.loc[x, "Hours"] > 12:
        df.drop(x, inplace=True)
print(df)
```

Discovering Duplicates

```python
# Topic: Discovering Duplicates
# duplicated() returns True for a row that is a copy of an earlier row
df = make_dirty()
print(df.duplicated())
```

Removing Duplicates

```python
# Topic: Removing Duplicates
# drop_duplicates() removes the repeated rows
df = make_dirty()
df.drop_duplicates(inplace=True)
print(df)
print("Rows left:", len(df))
```

All Cleaning Steps Together

```python
# Topic: All Cleaning Steps Together
# Now we clean the data fully, step by step
df = make_dirty()

df["Day"] = pd.to_datetime(df["Day"], format="mixed")   # fix wrong format
df.dropna(subset=["Day"], inplace=True)                  # remove row with no date
df.loc[df["Hours"] > 12, "Hours"] = 4                    # fix wrong data
df["Sleep"] = df["Sleep"].fillna(df["Sleep"].median())   # fill empty sleep
df["Score"] = df["Score"].fillna(df["Score"].mean())     # fill empty score
df.drop_duplicates(inplace=True)                         # remove duplicates
df.reset_index(drop=True, inplace=True)                  # renumber the rows

print(df)
print("Empty cells left:", df.isnull().sum().sum())
```
