What is a Series

```python
# Topic: Series
# A Series is a single column of data. It can hold any type of values.

import pandas as pd

temperatures = [31, 34, 29, 36]
s = pd.Series(temperatures)
print(s)
```

Labels

```python
# Topic: Labels (Index)
# By default each value gets a label starting from 0.
# We can use this label to get a value.

import pandas as pd

s = pd.Series([31, 34, 29, 36])
print("Value with label 0:", s[0])
print("Value with label 2:", s[2])
```

Create Labels

```python
# Topic: Custom Labels
# The index argument lets us give our own names to the labels

import pandas as pd

temperatures = pd.Series([31, 34, 29, 36], index=["Mon", "Tue", "Wed", "Thu"])
print(temperatures)

# Now we can use our own label to get a value
print("Temperature on Tue:", temperatures["Tue"])
```

Key/Value Objects as Series

```python
# Topic: Series from a Dictionary
# The keys of the dictionary become the labels

import pandas as pd

fruit_prices = {"apple": 300, "banana": 150, "mango": 400}
prices = pd.Series(fruit_prices)
print(prices)
```

Selecting Only Some Items

```python
# Topic: Choosing Items with index
# When creating from a dictionary, index can pick only the keys we want

import pandas as pd

fruit_prices = {"apple": 300, "banana": 150, "mango": 400}
some = pd.Series(fruit_prices, index=["apple", "mango"])
print(some)
```

DataFrames are Made of Series

```python
# Topic: From Series to DataFrame
# A DataFrame is a full table. Each column in it is a Series.

import pandas as pd

data = {
    "price": [300, 150, 400],
    "stock": [20, 35, 10]
}
df = pd.DataFrame(data, index=["apple", "banana", "mango"])
print(df)

# Taking one column out gives a Series
print(type(df["price"]))
```
