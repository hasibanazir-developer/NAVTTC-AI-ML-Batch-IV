Read CSV Files: Creating the Data

```python
# Topic: Creating a Sample CSV File
# Colab does not have a ready CSV file for us, so we first make one.
# Each row is one study session of a student.

import pandas as pd

study = {
    "Hours":    [2, 3, 1.5, 4, 2.5, 3.5, 1, 5, 2, 4.5, 3, 2, 6, 1.5, 3, 4, 2.5, 5.5, 3.5, 2],
    "Sleep":    [7, 6, 8, 6.5, 7, 6, 8, 5.5, 7.5, 6, 7, 8, 5, 7.5, 6.5, 6, 7, 5.5, 6.5, 7],
    "Breaks":   [2, 3, 1, 4, 2, 3, 1, 5, 2, 4, 3, 2, 5, 1, 3, 4, 2, 5, 3, 2],
    "Score":    [62, 70, 55, 82, 66, 76, 50, 91, 60, 86, 71, 63, 95, 57, 72, 80, 67, 90, 77, 61],
}
pd.DataFrame(study).to_csv("study.csv", index=False)
print("study.csv created")
```

Loading a CSV into a DataFrame

```python
# Topic: read_csv() and to_string()
# read_csv() loads the file. to_string() prints EVERY row of the table.

import pandas as pd

df = pd.read_csv("study.csv")
print(df.to_string())
```

Printing without to_string()

```python
# Topic: Printing a Large DataFrame
# Without to_string(), Pandas shows only the first and last rows
# when the table is bigger than the display limit

import pandas as pd

pd.options.display.max_rows = 10   # lower the limit to see the effect
df = pd.read_csv("study.csv")
print(df)
```

max_rows

```python
# Topic: max_rows Setting
# max_rows decides how many rows are shown before Pandas shortens the table

import pandas as pd

pd.reset_option("display.max_rows")   # go back to the default value
print("Default max_rows:", pd.options.display.max_rows)
```

Changing the Row Limit

```python
# Topic: Increasing max_rows
# If we raise the limit, the whole DataFrame is printed

import pandas as pd

pd.options.display.max_rows = 100
df = pd.read_csv("study.csv")
print(df)
```
