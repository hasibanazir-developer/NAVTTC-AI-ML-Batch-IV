Joining NumPy Arrays with concatenate()

```python
# Topic: concatenate()
# Joining means putting two or more arrays together into one

import numpy as np

morning = np.array([10, 12, 15])
evening = np.array([20, 18, 25])

all_day = np.concatenate((morning, evening))
print("Joined 1-D arrays:", all_day)
```

Joining 2-D Arrays along an Axis

```python
# Topic: concatenate() with axis
# axis=0 joins rows (one under another), axis=1 joins columns (side by side)

import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("axis=0 (add rows):")
print(np.concatenate((a, b), axis=0))

print("axis=1 (add columns):")
print(np.concatenate((a, b), axis=1))
```

Joining Arrays Using Stack Functions

```python
# Topic: stack()
# stack() is like concatenate, but it joins along a NEW axis

import numpy as np

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("stack with axis=0:")
print(np.stack((x, y), axis=0))

print("stack with axis=1:")
print(np.stack((x, y), axis=1))
```

Stacking Along Rows with hstack()

```python
# Topic: hstack()
# hstack() joins arrays horizontally (side by side)

import numpy as np

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(np.hstack((x, y)))
```

Stacking Along Columns with vstack()

```python
# Topic: vstack()
# vstack() joins arrays vertically (one on top of the other)

import numpy as np

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print(np.vstack((x, y)))
```

Stacking Along Height with dstack()

```python
# Topic: dstack()
# dstack() joins arrays along the depth (third) axis

import numpy as np

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

result = np.dstack((x, y))
print(result)
print("Shape:", result.shape)
```
