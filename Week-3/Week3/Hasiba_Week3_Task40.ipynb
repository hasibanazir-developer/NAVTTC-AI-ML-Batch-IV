What is a Random Number

```python
# Topic: Random Numbers
# Random means we cannot predict the value logically.
# Computers make "pseudo random" numbers using a formula.
# True random numbers need outside data, like mouse movements.
# A seed fixes the starting point, so the same "random" numbers appear every run.

import numpy as np

np.random.seed(7)
print("With seed 7:", np.random.randint(100, size=3))
np.random.seed(7)
print("Seed 7 again gives the same numbers:", np.random.randint(100, size=3))
```

Generate Random Number

```python
# Topic: randint()
# randint(n) gives a random whole number from 0 up to n-1

from numpy import random

dice = random.randint(1, 7)   # a dice roll from 1 to 6
print("Dice roll:", dice)

x = random.randint(100)       # 0 to 99
print("Random number below 100:", x)
```

Generate Random Float

```python
# Topic: rand()
# rand() gives a random decimal number between 0 and 1

from numpy import random

x = random.rand()
print("Random float:", x)
```

Generate Random Array of Integers

```python
# Topic: randint() with size
# The size argument creates a whole array of random integers

from numpy import random

one_d = random.randint(50, size=5)
print("1-D array of 5 random integers:", one_d)

two_d = random.randint(50, size=(3, 4))
print("2-D array (3 rows, 4 columns):")
print(two_d)
```

Generate Random Array of Floats

```python
# Topic: rand() with a shape
# Give the shape directly to rand() to get an array of random floats

from numpy import random

print("1-D array of 4 floats:", random.rand(4))
print("2-D array (2 rows, 3 columns):")
print(random.rand(2, 3))
```

Generate Random Number From Array

```python
# Topic: choice()
# choice() picks a random value from the values we give it

from numpy import random

colors = ["red", "green", "blue", "yellow"]
print("Random color:", random.choice(colors))

# size gives an array of random picks
print("Random 2-D array of choices:")
print(random.choice([2, 4, 6, 8], size=(3, 3)))
```
