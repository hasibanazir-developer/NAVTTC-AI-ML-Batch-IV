What is a DataFrame

```python
# Topic: DataFrame
# A DataFrame is a 2-D table with rows and columns, like an Excel sheet

import pandas as pd

employees = {
    "Name": ["Ali", "Sara", "Usman"],
    "Salary": [60000, 75000, 52000]
}

df = pd.DataFrame(employees)
print(df)
```

Locate Row

```python
# Topic: loc to Get Rows
# loc gets a row by its index label.
# One row gives a Series; a list of rows gives a DataFrame.

import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Usman"],
    "Salary": [60000, 75000, 52000]
})

print("Row 0 (a Series):")
print(df.loc[0])

print("Rows 0 and 2 (a DataFrame):")
print(df.loc[[0, 2]])
```

Named Indexes

```python
# Topic: Named Indexes
# Instead of 0, 1, 2 we can give our own names to the rows

import pandas as pd

df = pd.DataFrame({
    "Salary": [60000, 75000, 52000],
    "Department": ["IT", "HR", "Sales"]
}, index=["Ali", "Sara", "Usman"])

print(df)
```

Locate Named Indexes

```python
# Topic: loc with Named Indexes
# Put the row name inside loc to get that row

import pandas as pd

df = pd.DataFrame({
    "Salary": [60000, 75000, 52000],
    "Department": ["IT", "HR", "Sales"]
}, index=["Ali", "Sara", "Usman"])

print(df.loc["Sara"])
```

Load Files Into a DataFrame

```python
# Topic: Loading a File into a DataFrame
# If the data is saved in a CSV file, read_csv() turns it into a DataFrame

import pandas as pd

# First we save a small CSV file, then we read it back
pd.DataFrame({
    "Name": ["Ali", "Sara", "Usman"],
    "Salary": [60000, 75000, 52000]
}).to_csv("employees.csv", index=False)

df = pd.read_csv("employees.csv")
print(df)
```
