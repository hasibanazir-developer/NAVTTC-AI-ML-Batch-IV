Splitting NumPy Arrays

```python
# Topic: array_split()
# Splitting is the opposite of joining: one array is broken into parts

import numpy as np

week = np.array([5, 7, 6, 8, 4, 9])
parts = np.array_split(week, 3)

print("Split into 3 parts:", parts)
```

Splitting into Uneven Parts

```python
# Topic: Uneven Splits
# If the values cannot be shared equally, array_split() adjusts the last parts.
# split() gives an error in that case.

import numpy as np

week = np.array([5, 7, 6, 8, 4, 9])

print("Split into 4 parts:", np.array_split(week, 4))

try:
    np.split(week, 4)
except ValueError as error:
    print("split() error:", error)
```

Split Into Arrays

```python
# Topic: Accessing the Split Parts
# The result is a list of arrays, so we can use indexes

import numpy as np

week = np.array([5, 7, 6, 8, 4, 9])
parts = np.array_split(week, 3)

print("First part :", parts[0])
print("Second part:", parts[1])
print("Third part :", parts[2])
```

Splitting 2-D Arrays

```python
# Topic: Splitting a 2-D Array
# By default it splits by rows. axis=1 splits by columns.

import numpy as np

grid = np.arange(1, 19).reshape(6, 3)
print("Original 6 x 3 array:")
print(grid)

print("Split into 3 parts by rows:")
for part in np.array_split(grid, 3):
    print(part)

print("Split into 3 parts by columns (axis=1):")
for part in np.array_split(grid, 3, axis=1):
    print(part.ravel())
```

Alternate Solution: hsplit()

```python
# Topic: hsplit()
# hsplit() splits a 2-D array by columns (the opposite of hstack)

import numpy as np

grid = np.arange(1, 13).reshape(3, 4)
print(grid)

left, right = np.hsplit(grid, 2)
print("Left part:")
print(left)
print("Right part:")
print(right)
```

Similar Alternates: vsplit() and dsplit()

```python
# Topic: vsplit() and dsplit()
# vsplit() splits by rows (opposite of vstack)
# dsplit() splits by depth (opposite of dstack)

import numpy as np

grid = np.arange(1, 13).reshape(4, 3)
top, bottom = np.vsplit(grid, 2)
print("Top half:")
print(top)
print("Bottom half:")
print(bottom)

cube = np.arange(1, 9).reshape(2, 2, 2)
front, back = np.dsplit(cube, 2)
print("dsplit first part shape:", front.shape)
```
