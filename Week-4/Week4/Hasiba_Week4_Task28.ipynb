Importing NumPy

```python
# Topic: Importing NumPy
# NumPy is a Python library for working with arrays (fast lists of numbers).
# We import it with the short name np, which is the common practice.

import numpy as np

print("NumPy version:", np.__version__)
```

Create a NumPy ndarray Object

```python
# Topic: Creating an ndarray
# The array object in NumPy is called ndarray. We create it with np.array().

import numpy as np

# From a Python list
marks = np.array([78, 85, 92, 66])
print(marks)
print(type(marks))

# From a Python tuple
heights = np.array((5.2, 5.6, 6.0))
print(heights)
```

Dimensions in Arrays

```python
# Topic: Dimensions
# A dimension is one level of nesting inside an array.
# One list = 1 dimension, a list of lists = 2 dimensions, and so on.

import numpy as np

one_level = np.array([1, 2, 3])
two_levels = np.array([[1, 2, 3], [4, 5, 6]])

print("One level of brackets:", one_level)
print("Two levels of brackets:")
print(two_levels)
```

0-D Arrays

```python
# Topic: 0-D Array
# A 0-D array is just a single value (also called a scalar)

import numpy as np

temperature = np.array(37)
print(temperature)
print("Dimensions:", temperature.ndim)
```

1-D Arrays

```python
# Topic: 1-D Array
# A 1-D array is a simple row of values. This is the most common type.

import numpy as np

daily_steps = np.array([4500, 7200, 6100, 8000, 5300])
print(daily_steps)
print("Dimensions:", daily_steps.ndim)
```

2-D Arrays

```python
# Topic: 2-D Array
# A 2-D array is an array of 1-D arrays. It looks like a table or a matrix.

import numpy as np

# Each inner list is one student's marks in three subjects
class_marks = np.array([[80, 75, 90],
                        [65, 88, 72]])
print(class_marks)
print("Dimensions:", class_marks.ndim)
```

3-D Arrays

```python
# Topic: 3-D Array
# A 3-D array is an array of 2-D arrays.
# Example: marks of two classes, each class with two students and three subjects.

import numpy as np

school = np.array([
    [[80, 75, 90], [65, 88, 72]],   # class 1
    [[70, 60, 85], [95, 91, 89]]    # class 2
])
print(school)
print("Dimensions:", school.ndim)
```

Check Number of Dimensions

```python
# Topic: ndim
# The ndim attribute tells us how many dimensions an array has

import numpy as np

a = np.array(5)
b = np.array([1, 2, 3])
c = np.array([[1, 2], [3, 4]])
d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

for name, arr in [("a", a), ("b", b), ("c", c), ("d", d)]:
    print(name, "has", arr.ndim, "dimension(s)")
```

Higher Dimensional Arrays

```python
# Topic: Higher Dimensions with ndmin
# ndmin lets us choose the minimum number of dimensions while creating an array

import numpy as np

arr = np.array([10, 20, 30], ndmin=4)
print(arr)
print("Number of dimensions:", arr.ndim)
# The innermost dimension has 3 values; the other 3 dimensions have 1 value each
print("Shape:", arr.shape)
```
