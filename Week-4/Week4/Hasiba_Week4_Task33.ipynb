Shape of an Array

```python
# Topic: Shape
# The shape tells how many values are in each dimension

import numpy as np

week_temps = np.array([30, 32, 31, 29, 33, 34, 28])
print("Values:", week_temps)
print("Shape:", week_temps.shape)   # (7,) means 7 values in one dimension
```

Get the Shape of an Array

```python
# Topic: shape Attribute
# shape gives a tuple: (rows, columns) for a 2-D array

import numpy as np

# 3 students, 4 subjects each
marks = np.array([[80, 75, 90, 66],
                  [65, 88, 72, 91],
                  [70, 60, 85, 77]])
print(marks)
print("Shape:", marks.shape)
print("Rows:", marks.shape[0], "| Columns:", marks.shape[1])
```

Shape of a Higher Dimensional Array

```python
# Topic: Shape with ndmin
# Here we make a 5-D array from a list of 4 values

import numpy as np

arr = np.array([1, 2, 3, 4], ndmin=5)
print(arr)
print("Shape:", arr.shape)
```

What does the Shape Tuple Represent

```python
# Topic: Reading the Shape Tuple
# Each number in the shape tuple is the size of one dimension

import numpy as np

arr = np.array([1, 2, 3, 4], ndmin=5)

for position, size in enumerate(arr.shape):
    print("Dimension", position, "has", size, "element(s)")
```
