Difference Between Copy and View

```python
# Topic: Copy vs View
# A copy is a brand new array with its own data.
# A view is just another way to look at the SAME data as the original.

import numpy as np

original = np.array([10, 20, 30, 40])
c = original.copy()
v = original.view()

print("Original:", original)
print("Copy    :", c)
print("View    :", v)
```

Copy

```python
# Topic: Copy
# Changing the original array does NOT change the copy

import numpy as np

scores = np.array([50, 60, 70, 80])
backup = scores.copy()

scores[0] = 99

print("Original after change:", scores)
print("Copy stays the same  :", backup)
```

View

```python
# Topic: View
# Changing the original array DOES change the view

import numpy as np

scores = np.array([50, 60, 70, 80])
window = scores.view()

scores[0] = 99

print("Original after change:", scores)
print("View also changed    :", window)
```

Make Changes in the View

```python
# Topic: Changing the View
# Changing the view also changes the original array

import numpy as np

scores = np.array([50, 60, 70, 80])
window = scores.view()

window[3] = 100

print("View after change    :", window)
print("Original also changed:", scores)

# Slicing also gives a view
part = scores[1:3]
part[0] = 0
print("After changing a slice, original is:", scores)
```

Check if Array Owns its Data

```python
# Topic: base Attribute
# base is None if the array owns its data (a copy).
# base points to the original array if it is a view.

import numpy as np

arr = np.array([1, 2, 3, 4, 5])
c = arr.copy()
v = arr.view()

print("Copy base:", c.base)
print("View base:", v.base)
```
