Iterating Arrays

```python
# Topic: Iterating a 1-D Array
# Iterating means going through the values one by one with a for loop

import numpy as np

speeds = np.array([40, 55, 60, 72])

for s in speeds:
    print("Speed:", s, "km/h")
```

Iterating 2-D Arrays

```python
# Topic: Iterating a 2-D Array
# A loop over a 2-D array gives one row at a time.
# A nested loop gives each single value.

import numpy as np

table = np.array([[1, 2, 3],
                  [4, 5, 6]])

for row in table:
    print("Row:", row)

for row in table:
    for value in row:
        print(value, end=" ")
print()
```

Iterating 3-D Arrays

```python
# Topic: Iterating a 3-D Array
# A loop over a 3-D array gives one 2-D block at a time.
# Three nested loops reach each single value.

import numpy as np

cube = np.array([[[1, 2], [3, 4]],
                 [[5, 6], [7, 8]]])

for block in cube:
    print("Block:")
    print(block)

for block in cube:
    for row in block:
        for value in row:
            print(value, end=" ")
print()
```

Iterating Arrays Using nditer()

```python
# Topic: nditer()
# nditer() visits every single value, no matter how many dimensions,
# so we do not need many nested loops

import numpy as np

cube = np.array([[[1, 2], [3, 4]],
                 [[5, 6], [7, 8]]])

for value in np.nditer(cube):
    print(value, end=" ")
print()
```

Iterating Array With Different Data Types

```python
# Topic: Changing Type While Iterating
# op_dtypes changes the type of each value during the loop.
# NumPy needs extra space for this, so we add flags=['buffered'].

import numpy as np

nums = np.array([1, 2, 3])

for value in np.nditer(nums, flags=["buffered"], op_dtypes=["S"]):
    print(value)
```

Iterating With Different Step Size

```python
# Topic: Skipping Values While Iterating
# We slice first, then iterate, to skip some values

import numpy as np

grid = np.array([[1, 2, 3, 4],
                 [5, 6, 7, 8]])

# [:, ::2] keeps every row but only every second column
for value in np.nditer(grid[:, ::2]):
    print(value, end=" ")
print()
```

Enumerated Iteration Using ndenumerate()

```python
# Topic: ndenumerate()
# ndenumerate() gives the index (position) and the value together

import numpy as np

fruits = np.array(["apple", "mango", "kiwi"])
for index, fruit in np.ndenumerate(fruits):
    print(index, fruit)

grid = np.array([[10, 20],
                 [30, 40]])
for index, value in np.ndenumerate(grid):
    print("Position", index, "has value", value)
```
