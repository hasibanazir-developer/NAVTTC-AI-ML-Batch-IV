Creating the Data

```python
# Topic: Creating a Sample CSV File
# Colab does not have a ready CSV file for us, so we first make one.
# Each row is one study session of a student.

import pandas as pd

study = {
    "Hours":    [2, 3, 1.5, 4, 2.5, 3.5, 1, 5, 2, 4.5, 3, 2, 6, 1.5, 3, 4, 2.5, 5.5, 3.5, 2],
    "Sleep":    [7, 6, 8, 6.5, 7, 6, 8, 5.5, 7.5, 6, 7, 8, 5, 7.5, 6.5, 6, 7, 5.5, 6.5, 7],
    "Breaks":   [2, 3, 1, 4, 2, 3, 1, 5, 2, 4, 3, 2, 5, 1, 3, 4, 2, 5, 3, 2],
    "Score":    [62, 70, 55, 82, 66, 76, 50, 91, 60, 86, 71, 63, 95, 57, 72, 80, 67, 90, 77, 61],
}
pd.DataFrame(study).to_csv("study.csv", index=False)
print("study.csv created")
```

Finding Relationships with corr()

```python
# Topic: corr()
# corr() shows how strongly each number column is related to the others

import pandas as pd

df = pd.read_csv("study.csv")
print(df.corr().round(2))
```

Result Explained

```python
# Topic: Reading the Correlation Values
# Values go from -1 to 1:
#  1   -> perfect relationship (both go up together)
#  0.6 or more -> good relationship
# -1   -> perfect opposite relationship (one goes up, the other goes down)
#  near 0 -> no useful relationship

import pandas as pd

df = pd.read_csv("study.csv")
corr = df.corr()

def describe(value):
    if value >= 0.6:
        return "good positive relationship"
    if value <= -0.6:
        return "good negative relationship"
    return "weak relationship"

for col in ["Sleep", "Breaks", "Score"]:
    v = corr.loc["Hours", col]
    print(f"Hours vs {col}: {v:.2f} -> {describe(v)}")
```

Perfect Correlation

```python
# Topic: Perfect Correlation
# Every column has a correlation of exactly 1 with itself

import pandas as pd

df = pd.read_csv("study.csv")
corr = df.corr()

for col in df.columns:
    print(col, "vs", col, "=", corr.loc[col, col])
```

Good Correlation

```python
# Topic: Good Correlation
# Hours and Score have a high value (close to 1),
# so more study hours usually means a higher score.
# We could use Hours to predict Score.

import pandas as pd

df = pd.read_csv("study.csv")
print("Hours vs Score:", round(df["Hours"].corr(df["Score"]), 2))

# Sleep goes DOWN when Hours goes UP, so it has a strong negative value
print("Hours vs Sleep:", round(df["Hours"].corr(df["Sleep"]), 2))
```

Bad Correlation

```python
# Topic: Bad Correlation
# Here we add a column of random lucky numbers.
# It has nothing to do with study hours, so the value is close to 0
# and it cannot be used for prediction.

import pandas as pd
import numpy as np

df = pd.read_csv("study.csv")
np.random.seed(25)
df["LuckyNumber"] = np.random.randint(1, 100, size=len(df))

print("Hours vs LuckyNumber:", round(df["Hours"].corr(df["LuckyNumber"]), 2))
```
