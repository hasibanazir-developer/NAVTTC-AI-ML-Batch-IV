Searching Arrays with where()

```python
# Topic: where()
# where() returns the positions (indexes) where a condition is True

import numpy as np

answers = np.array(["A", "C", "B", "A", "D", "A"])
positions = np.where(answers == "A")

print("Answer A is at positions:", positions)
```

Finding Even and Odd Values

```python
# Topic: where() with Conditions
# % 2 == 0 means even, % 2 == 1 means odd

import numpy as np

nums = np.array([3, 8, 11, 14, 17, 20])

print("Even values are at:", np.where(nums % 2 == 0))
print("Odd values are at:", np.where(nums % 2 == 1))

# where() can also choose between two results for every value
print(np.where(nums % 2 == 0, "even", "odd"))
```

Search Sorted

```python
# Topic: searchsorted()
# searchsorted() works on a SORTED array.
# It tells where a value should be inserted to keep the array in order.

import numpy as np

sorted_scores = np.array([40, 55, 70, 85])
index = np.searchsorted(sorted_scores, 60)

print("60 should be inserted at index", index)
```

Search From the Right Side

```python
# Topic: searchsorted() with side='right'
# When the value already exists, side='right' gives the position after it

import numpy as np

arr = np.array([10, 20, 20, 30])

print("Left side index for 20 :", np.searchsorted(arr, 20))
print("Right side index for 20:", np.searchsorted(arr, 20, side="right"))
```

Multiple Values

```python
# Topic: Searching Many Values at Once
# Pass a list of values to get a list of insertion positions

import numpy as np

sorted_scores = np.array([40, 55, 70, 85])
new_scores = [35, 60, 90]

print("Insert positions:", np.searchsorted(sorted_scores, new_scores))
```
