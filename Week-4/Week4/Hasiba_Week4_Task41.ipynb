Why Use Pandas

```python
# Topic: Why Use Pandas
# Pandas makes big data easy to read and work with.
# Here a messy dictionary becomes a clean, readable table in one line.

import pandas as pd

shop = {
    "Item": ["Tea", "Sugar", "Milk", "Bread"],
    "Price": [450, 160, 220, 180],
    "Quantity": [2, 1, 3, 2]
}

table = pd.DataFrame(shop)
print(table)
```

What Can Pandas Do

```python
# Topic: What Pandas Can Do
# Pandas can answer questions about data, for example:
# the average, the maximum, the minimum, and which columns are related.

import pandas as pd

shop = pd.DataFrame({
    "Item": ["Tea", "Sugar", "Milk", "Bread"],
    "Price": [450, 160, 220, 180],
    "Quantity": [2, 1, 3, 2]
})

shop["Total"] = shop["Price"] * shop["Quantity"]   # add a new column

print(shop)
print("Average price:", shop["Price"].mean())
print("Most expensive item price:", shop["Price"].max())
print("Cheapest item price:", shop["Price"].min())
print("Total bill:", shop["Total"].sum())
```

Removing Empty Values

```python
# Topic: Cleaning Data with Pandas
# Pandas can find and remove rows that have missing values

import pandas as pd
import numpy as np

shop = pd.DataFrame({
    "Item": ["Tea", "Sugar", "Milk", "Bread"],
    "Price": [450, np.nan, 220, 180]   # np.nan means a missing value
})

print("Before cleaning:")
print(shop)
print("After removing the empty row:")
print(shop.dropna())
```

Pandas Basic Example: Reading a CSV

```python
# Topic: Loading a CSV into a DataFrame
# The most common way to start with Pandas is to read a CSV file

import pandas as pd

pd.DataFrame({"City": ["Lahore", "Karachi", "Islamabad"],
              "Population_millions": [13.0, 16.8, 1.2]}).to_csv("cities.csv", index=False)

df = pd.read_csv("cities.csv")
print(df.to_string())
```
