Creating the Data

```python
# Topic: Creating a Sample CSV File
# Colab does not have a ready CSV file for us, so we first make one.
# Each row is one study session of a student.

import pandas as pd

study = {
    "Hours":    [2, 3, 1.5, 4, 2.5, 3.5, 1, 5, 2, 4.5, 3, 2, 6, 1.5, 3, 4, 2.5, 5.5, 3.5, 2],
    "Sleep":    [7, 6, 8, 6.5, 7, 6, 8, 5.5, 7.5, 6, 7, 8, 5, 7.5, 6.5, 6, 7, 5.5, 6.5, 7],
    "Breaks":   [2, 3, 1, 4, 2, 3, 1, 5, 2, 4, 3, 2, 5, 1, 3, 4, 2, 5, 3, 2],
    "Score":    [62, 70, 55, 82, 66, 76, 50, 91, 60, 86, 71, 63, 95, 57, 72, 80, 67, 90, 77, 61],
}
pd.DataFrame(study).to_csv("study.csv", index=False)
print("study.csv created")
```

Viewing the Data with head()

```python
# Topic: head()
# head() shows the first rows of the table (5 rows if we give no number)

import pandas as pd

df = pd.read_csv("study.csv")

print("First 5 rows:")
print(df.head())

print("First 8 rows:")
print(df.head(8))
```

Viewing the Data with tail()

```python
# Topic: tail()
# tail() shows the last rows of the table

import pandas as pd

df = pd.read_csv("study.csv")

print("Last 5 rows:")
print(df.tail())
```

Info About the Data

```python
# Topic: info()
# info() gives a short summary: number of rows, column names,
# how many values are filled (non-null) and the data type of each column

import pandas as pd

df = pd.read_csv("study.csv")
df.info()
```

Null Values

```python
# Topic: Finding Empty (Null) Values
# Here we make some values empty on purpose.
# info() then shows fewer non-null values in those columns.

import pandas as pd
import numpy as np

df = pd.read_csv("study.csv")
df.loc[[3, 9, 15], "Score"] = np.nan   # remove 3 scores
df.loc[6, "Sleep"] = np.nan            # remove 1 sleep value

df.info()

# isnull().sum() counts empty values in each column
print("\nEmpty values in each column:")
print(df.isnull().sum())
print("\nEmpty rows should be cleaned before analysis (see the next task).")
```

Quick Statistics with describe()

```python
# Topic: describe()
# describe() gives count, mean, min, max and more for each number column

import pandas as pd

df = pd.read_csv("study.csv")
print(df.describe())
```
