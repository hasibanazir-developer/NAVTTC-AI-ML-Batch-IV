```python
# Task 5: Program to calculate the square root of a number

# Store the number whose square root we want
# Change this value to get a different result
num = 8

# Uncomment the line below to take the number from the user instead
# num = float(input('Enter a number: '))

# Raising a number to the power 0.5 gives its square root
num_sqrt = num ** 0.5

# Display the result
# %0.3f means show the number with 3 digits after the decimal point
print('The square root of %0.3f is %0.3f' % (num, num_sqrt))
```
