```python
# Task 4: Program to add two numbers entered by the user

# Take the first number from the user
# input() always returns the value as a string (text)
num1 = input('Enter first number: ')

# Take the second number from the user
num2 = input('Enter second number: ')

# Convert both strings into float (decimal numbers) and add them
# Without float(), Python would join the strings instead of adding them
sum = float(num1) + float(num2)

# Display the result
print('The sum of {0} and {1} is {2}'.format(num1, num2, sum))
```
