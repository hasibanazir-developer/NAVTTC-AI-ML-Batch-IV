Using enumerate()

```python
# Topic: enumerate()
# enumerate() gives us the position (index) and the value together while looping

fruits = ["mango", "apple", "orange", "grapes"]

for index, fruit in enumerate(fruits):
    print("Index", index, "->", fruit)
```

Using zip()

```python
# Topic: zip()
# zip() lets us loop over two (or more) lists at the same time.
# It pairs items by position and stops at the shorter list.

subjects = ["Math", "English", "Physics"]
scores = [88, 92, 79]

for subject, score in zip(subjects, scores):
    print(subject, "score:", score)

# zip() stops at the shorter list, so "Delhi" is ignored here
countries = ["Pakistan", "Turkey"]
capitals = ["Islamabad", "Ankara", "Delhi"]
for country, capital in zip(countries, capitals):
    print("The capital of", country, "is", capital)
```

Using items()

```python
# Topic: items()
# items() gives us both the key and the value of a dictionary while looping

prices = {"tea": 80, "coffee": 250, "juice": 150}

for item, price in prices.items():
    print(item, "costs Rs.", price)
```

Using sorted()

```python
# Topic: sorted()
# sorted() lets us loop through items in order (smallest to largest)
# without changing the original list

ages = [25, 18, 32, 18, 21, 25]

print("Sorted ages:")
for age in sorted(ages):
    print(age, end=" ")
print()

# Using set() inside sorted() removes duplicate values first
print("Sorted unique ages:")
for age in sorted(set(ages)):
    print(age, end=" ")
print()

print("Original list is still the same:", ages)
```

Using reversed()

```python
# Topic: reversed()
# reversed() lets us loop from the last item to the first item
# without changing the original list

steps = [1, 2, 3, 4, 5]

for s in reversed(steps):
    print(s, end=" ")
print()

print("Original list:", steps)
```

While Loop with If Condition

```python
# Topic: While Loop with an If Condition
# A while loop counts up and does something special at one value

counter = 1

while counter <= 6:
    if counter == 4:
        print("Counter reached 4, halfway done!")
    else:
        print("Counter:", counter)
    counter += 1
```
