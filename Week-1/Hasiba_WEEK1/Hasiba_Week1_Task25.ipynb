Creating a Dictionary

```python
# Topic: Creating a Dictionary
# A dictionary stores data as key: value pairs inside curly brackets

student = {
    "name": "Haseeba",
    "course": "AI",
    "week": 1
}
print(student)

# dict() is another way to create a dictionary
phone = dict(brand="Samsung", year=2024)
print(phone)
```

Valid and Invalid Dictionary Keys

```python
# Topic: Valid and Invalid Keys
# Keys must be unchangeable types like strings, numbers or tuples.
# Values can be anything, even lists.

valid = {
    1: "one",
    "two": 2,
    (3, 4): "a tuple key",
    "list_value": [5, 6, 7]
}
print(valid)

# A list cannot be a key because lists can change
try:
    invalid = {[1, 2]: "not allowed"}
except TypeError as error:
    print("Error:", error)
```

Keys Must Be Unique

```python
# Topic: Unique Keys
# If the same key is written twice, the last value replaces the first one

capitals = {
    "Punjab": "Multan",
    "Sindh": "Karachi",
    "Punjab": "Lahore"
}
print(capitals)
```

Accessing Dictionary Items

```python
# Topic: Accessing Items
# Use the key in square brackets, or the get() method

book = {"title": "Python Basics", "pages": 220}

print("Title:", book["title"])
print("Pages:", book.get("pages"))

# get() is safer: it gives None (or a default) instead of an error
print("Author:", book.get("author"))
print("Author:", book.get("author", "Unknown"))

try:
    print(book["author"])
except KeyError as error:
    print("Error with []:", error)
```

Adding Items to a Dictionary

```python
# Topic: Adding Items
# Assign a value to a new key to add it

profile = {"name": "Haseeba"}
profile["city"] = "Lahore"
profile["age"] = 21
print(profile)
```

Removing Dictionary Items

```python
# Topic: Removing Items
# del removes a key, pop() removes a key and gives back its value,
# popitem() removes the last added item, and clear() removes everything

menu = {"burger": 500, "pizza": 1200, "fries": 250, "drink": 150}

del menu["fries"]
print("After del:", menu)

price = menu.pop("pizza")
print("Removed pizza costing", price, "| Menu now:", menu)

last_item = menu.popitem()
print("popitem() removed:", last_item, "| Menu now:", menu)

menu.clear()
print("After clear:", menu)
```

Changing Dictionary Items

```python
# Topic: Changing Items
# Assign a new value to an existing key, or use update()

car = {"brand": "Toyota", "model": "Corolla", "year": 2020}

car["year"] = 2024
print("After changing year:", car)

# update() can change existing keys and add new ones together
car.update({"model": "Yaris", "color": "white"})
print("After update():", car)
```

Iterating Through a Dictionary

```python
# Topic: Looping Through a Dictionary
# Since Python 3.7, dictionaries remember the order in which items were added

marks = {"Math": 88, "English": 76, "Computer": 95}

# Loop over keys
for subject in marks:
    print("Subject:", subject)

# Loop over values
for score in marks.values():
    print("Score:", score)

# Loop over keys and values together
for subject, score in marks.items():
    print(subject, "->", score)
```

Finding Dictionary Length

```python
# Topic: len()
# len() tells how many key-value pairs the dictionary has

marks = {"Math": 88, "English": 76, "Computer": 95}
print("Number of subjects:", len(marks))
```

Dictionary Methods: keys(), values(), copy()

```python
# Topic: More Dictionary Methods

fees = {"Python": 5000, "ML": 8000, "DL": 10000}

print("Keys:", list(fees.keys()))
print("Values:", list(fees.values()))
print("Items:", list(fees.items()))

# copy() makes a separate copy, so changes do not affect the original
backup = fees.copy()
backup["Python"] = 0
print("Original:", fees)
print("Copy:", backup)
```

Dictionary Membership Test

```python
# Topic: Membership Test
# 'in' and 'not in' check the KEYS of a dictionary, not the values

stock = {"pens": 40, "notebooks": 25}

print("Is 'pens' a key?", "pens" in stock)
print("Is 'erasers' not a key?", "erasers" not in stock)
print("Is 40 a key?", 40 in stock)   # False, because 40 is a value
```

Nested Dictionary and Dictionary Comprehension

```python
# Topic: Nested Dictionary and Dictionary Comprehension

# A nested dictionary has dictionaries inside it
team = {
    "member1": {"name": "Ali", "role": "Data Analyst"},
    "member2": {"name": "Haseeba", "role": "ML Engineer"}
}
for key, info in team.items():
    print(key, "-", info["name"], "works as", info["role"])

# Dictionary comprehension builds a dictionary in one line
cubes = {n: n ** 3 for n in range(1, 6)}
print("Cubes:", cubes)
```
