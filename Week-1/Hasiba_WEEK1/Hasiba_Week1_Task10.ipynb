```python
# Task 10: Program to convert kilometers into miles

# Take the distance in kilometers from the user and convert it into a float
kilometers = float(input("Enter value in kilometers: "))

# 1 kilometer is equal to 0.621371 miles
conv_fac = 0.621371

# Multiply kilometers by the conversion factor to get miles
miles = kilometers * conv_fac

# Display the result with 2 digits after the decimal point
print('%0.2f kilometers is equal to %0.2f miles' % (kilometers, miles))
```
