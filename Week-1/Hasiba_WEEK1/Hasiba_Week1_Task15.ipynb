Basic For Loop (Iterating over a List)

```python
# Topic: Basic For Loop
# A for loop is used to go through each item of a sequence (list, tuple, string, range) one by one.

# Create a list of city names
cities = ["Lahore", "Karachi", "Islamabad", "Multan"]

# The loop picks one city at a time, stores it in the variable 'city',
# and runs the print statement for each city
for city in cities:
    print("City name:", city)
```

For Loop over a Tuple

```python
# Topic: For Loop with a Tuple
# A for loop works on tuples in the same way it works on lists

# Create a tuple of subject names
subjects = ("Python", "Statistics", "Machine Learning")

# Print each subject one by one
for subject in subjects:
    print("I am learning", subject)
```

Loop Through a String

```python
# Topic: Loop Through a String
# A string is a sequence of characters, so a for loop can visit each character one by one

# Store a word in a variable
word = "Python"

# The loop takes one character at a time and prints it
for letter in word:
    print(letter)
```

range() with One Argument

```python
# Topic: range(stop)
# range(stop) creates numbers starting from 0 up to stop-1

# This prints numbers 0, 1, 2, 3, 4 (5 is not included)
for num in range(5):
    print(num)
```

range() with Two Arguments

```python
# Topic: range(start, stop)
# range(start, stop) creates numbers from start up to stop-1

# This prints numbers from 1 to 10
for num in range(1, 11):
    print(num)
```

range() with Three Arguments

```python
# Topic: range(start, stop, step)
# range(start, stop, step) creates numbers from start to stop-1,
# increasing by 'step' each time

# Print odd numbers from 1 to 19 (step of 2)
for num in range(1, 20, 2):
    print(num)

# A negative step makes the loop count backwards
# Print a countdown from 10 to 1
for num in range(10, 0, -1):
    print("Countdown:", num)
```

Continue Statement

```python
# Topic: Continue Statement
# 'continue' skips the rest of the code for the current item
# and moves directly to the next item in the loop

# Print every letter of the word except vowels
for ch in "programming":
    # If the letter is a vowel, skip it
    if ch in "aeiou":
        continue
    print(ch)
```

Break Statement

```python
# Topic: Break Statement
# 'break' stops the loop completely as soon as a condition becomes true

# Search for the first number divisible by 7 in a list
numbers = [3, 10, 15, 21, 28, 35]

for n in numbers:
    print("Checking:", n)
    # When a number divisible by 7 is found, stop the loop
    if n % 7 == 0:
        print("First number divisible by 7 is", n)
        break
```

Pass Statement

```python
# Topic: Pass Statement
# 'pass' does nothing. It is used as a placeholder when Python
# needs a statement but we do not want to write any code yet.

# Loop through numbers 1 to 5 without doing anything inside the loop
for i in range(1, 6):
    pass

# After the loop, i holds the last value from the range
print("Loop finished. Last value of i is", i)

# pass inside an if block: do nothing for even numbers, print odd numbers
for i in range(1, 6):
    if i % 2 == 0:
        pass   # placeholder, nothing happens here
    else:
        print("Odd number:", i)
```

Else with For Loop

```python
# Topic: Else Statement with For Loop
# The else block runs only if the loop finishes normally (without a break)

# Example 1: loop completes normally, so else block runs
for i in range(1, 4):
    print("Value:", i)
else:
    print("Loop completed without break")

# Example 2: loop is stopped by break, so else block does NOT run
for i in range(1, 6):
    if i == 3:
        print("Break at", i)
        break
    print("Value:", i)
else:
    print("This line will not be printed")
```

Using enumerate()

```python
# Topic: enumerate()
# enumerate() gives both the index (position) and the value of each item in the loop

# Create a list of tasks
tasks = ["wake up", "study", "practice code"]

# 'index' stores the position and 'task' stores the item
for index, task in enumerate(tasks):
    print(index, task)

# enumerate() can also start counting from a different number, like 1
for number, task in enumerate(tasks, start=1):
    print("Task", number, ":", task)
```

Nested For Loops

```python
# Topic: Nested For Loops
# A nested loop is a loop inside another loop.
# The inner loop runs completely for every single value of the outer loop.

# Print all combinations of rows and columns
for row in range(1, 3):
    for col in range(1, 4):
        print("Row", row, "Column", col)

# Print multiplication tables of 2 and 3
for table in range(2, 4):
    print("Table of", table)
    for i in range(1, 6):
        print(table, "x", i, "=", table * i)
```

Nested Loop Pattern (Practice)

```python
# Topic: Nested For Loop Practice - Star Pattern
# The outer loop controls the rows and the inner loop prints stars in each row

rows = 5
for i in range(1, rows + 1):
    # Print i stars in the current row on the same line
    for j in range(i):
        print("*", end=" ")
    # Move to the next line after each row
    print()
```
