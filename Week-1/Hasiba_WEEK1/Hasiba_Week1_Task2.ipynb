```python
# Task 2: Linux Commands
# In Colab, shell commands are run by adding "!" before them

# Show current working directory
!pwd

# Create a new folder and move into it
!mkdir -p myfolder
%cd myfolder

# List files in detail
!ls -l

# Create an empty file
!touch file1.txt

# Redirection: write (>) and append (>>) text to a file
!echo "Hello Linux" > file1.txt
!echo "Second line" >> file1.txt

# Read file contents
!cat file1.txt
!more file1.txt
!less file1.txt | head
!head -n 1 file1.txt

# Show file type
!file file1.txt

# Copy, move/rename and remove files
!cp file1.txt file2.txt
!mv file2.txt file3.txt
!ls
!rm file3.txt

# Create and remove an empty directory
!mkdir emptydir
!rmdir emptydir

# Change file permissions
!chmod 755 file1.txt
!ls -l

# Create an alias (shortcut for a command)
!alias ll='ls -l'

# Run a command with admin rights
!sudo echo "sudo works"

# Run commands using bash and sh shells
!bash -c "echo Running with bash"
!sh -c "echo Running with sh"

# Show running processes
!ps

# Start a background process and then kill it
!sleep 100 &
!pkill sleep

# List installed packages
!dpkg -l | head -n 5

# Show manual page of a command
!man ls | head -n 10

# The following commands do not work on Colab, so they are written as comments:
# sudo shutdown now   -> shuts down the system
# sudo reboot         -> restarts the system
# nano file1.txt      -> opens a text editor

# Go back to the previous folder
%cd ..
```
