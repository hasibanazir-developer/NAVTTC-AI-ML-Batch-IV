```python
# Task 3: Program to add two numbers

# Store the first number in a variable
num1 = 1.5

# Store the second number in a variable
num2 = 6.3

# Add both numbers and store the result in a variable called sum
sum = num1 + num2

# Display the result using format()
# {0}, {1} and {2} are replaced by num1, num2 and sum in order
print('The sum of {0} and {1} is {2}'.format(num1, num2, sum))
```
