```python
# Task 12: Program to show how an if-else statement works

# Store the value 20 in variable i
i = 20

# Check if i is smaller than 15
if (i < 15):
    # This block runs only when the condition is True
    print("i is smaller than 15")
    print("i'm in if Block")
else:
    # This block runs when the condition is False
    # Since 20 is not smaller than 15, this block will run
    print("i is greater than 15")
    print("i'm in else Block")

# This line is outside both blocks, so it always runs
print("i'm not in if and not in else Block")
```
