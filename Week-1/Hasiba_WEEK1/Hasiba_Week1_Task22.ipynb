Creating a Tuple

```python
# Topic: Creating a Tuple
# A tuple is like a list, but its items cannot be changed after creation.
# Tuples are written with round brackets ().

empty_tuple = ()
numbers = (10, 20, 30)
mixed = ("Haseeba", 21, 3.8)
nested = ("AI", [1, 2, 3], (4, 5))

print(empty_tuple)
print(numbers)
print(mixed)
print(nested)
```

Tuple with One Item

```python
# Topic: Single Item Tuple
# For one item, we must add a comma, otherwise Python treats it as a normal value

not_a_tuple = ("hello")
real_tuple = ("hello",)

print(type(not_a_tuple))   # str
print(type(real_tuple))    # tuple

# Brackets are optional; a comma alone also creates a tuple
packed = 1, 2, 3
print(packed, type(packed))
```

Accessing Items: Indexing

```python
# Topic: Indexing
# Each item has an index starting from 0

colors = ("red", "green", "blue", "yellow")

print("First color:", colors[0])
print("Third color:", colors[2])

# Nested tuple: use two indexes
info = ("Ali", (90, 85, 77))
print("Second mark of Ali:", info[1][1])

# Using an index that does not exist gives an error
try:
    print(colors[10])
except IndexError as error:
    print("Error:", error)
```

Accessing Items: Negative Indexing

```python
# Topic: Negative Indexing
# -1 is the last item, -2 is the second last, and so on

colors = ("red", "green", "blue", "yellow")

print("Last color:", colors[-1])
print("Second last color:", colors[-2])
```

Accessing Items: Slicing

```python
# Topic: Slicing
# tuple[start:stop:step] gives a part of the tuple

months = ("Jan", "Feb", "Mar", "Apr", "May", "Jun")

print(months[1:4])    # Feb to Apr
print(months[:2])     # first two months
print(months[3:])     # from Apr to the end
print(months[::2])    # every second month
```

Changing or Deleting a Tuple

```python
# Topic: Tuples Cannot Be Changed
# We cannot change or remove a single item of a tuple

scores = (50, 60, 70)

try:
    scores[0] = 99
except TypeError as error:
    print("Error:", error)

# But if a tuple contains a list, that list can still be changed
data = ("marks", [1, 2, 3])
data[1].append(4)
print("Tuple with changed list:", data)

# We can reassign the whole tuple
scores = (80, 90)
print("Reassigned tuple:", scores)

# We can delete the whole tuple using del
del scores
print("Tuple deleted")
```

Tuple Methods: count() and index()

```python
# Topic: Tuple Methods
# count() tells how many times a value appears
# index() tells the position of the first match

grades = ("A", "B", "A", "C", "A", "B")

print("Count of A:", grades.count("A"))
print("First position of C:", grades.index("C"))
```

Tuple Membership Test

```python
# Topic: Membership Test
# 'in' and 'not in' check if a value exists in the tuple

weekend = ("Saturday", "Sunday")

print("Is Sunday a weekend day?", "Sunday" in weekend)
print("Is Monday not a weekend day?", "Monday" not in weekend)
```

Iterating Through a Tuple

```python
# Topic: Iterating Through a Tuple
# A for loop visits each item one by one

planets = ("Mercury", "Venus", "Earth")

for planet in planets:
    print("Planet:", planet)
```

Built-in Functions with Tuple

```python
# Topic: Built-in Functions with Tuples

temps = (31, 27, 35, 29, 33)

print("Number of items:", len(temps))
print("Lowest temperature:", min(temps))
print("Highest temperature:", max(temps))
print("Sum of all:", sum(temps))
print("Sorted (gives a list):", sorted(temps))
```

Advantage of Tuple over List

```python
# Topic: Why Use a Tuple Instead of a List
# 1. Tuples protect data from accidental changes
# 2. Tuples are slightly faster than lists
# 3. Tuples can be used as dictionary keys, lists cannot

locations = {
    (31.52, 74.35): "Lahore",
    (24.86, 67.00): "Karachi"
}
print("City at (31.52, 74.35):", locations[(31.52, 74.35)])

# Trying a list as a key gives an error
try:
    bad = {[1, 2]: "test"}
except TypeError as error:
    print("Error with list as key:", error)
```
