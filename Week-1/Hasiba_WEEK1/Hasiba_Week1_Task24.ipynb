Creating a Set

```python
# Topic: Creating a Set
# A set holds unique items inside curly brackets {}. Its order is not fixed.

numbers = {1, 2, 3, 4}
mixed = {"AI", 10, 3.5, (1, 2)}   # different types are allowed

print(numbers)
print(mixed)

# Lists cannot be put inside a set because they can change
try:
    bad = {1, [2, 3]}
except TypeError as error:
    print("Error:", error)
```

Creating an Empty Set

```python
# Topic: Empty Set
# {} creates an empty DICTIONARY, so we use set() for an empty set

a = {}
b = set()

print(type(a))   # dict
print(type(b))   # set
```

Duplicate Items in a Set

```python
# Topic: No Duplicates
# A set automatically removes repeated values

visitors = {"Ali", "Sara", "Ali", "Zain", "Sara"}
print(visitors)

# This is a quick way to remove duplicates from a list
ids = [101, 102, 101, 103, 102]
print("Unique ids:", set(ids))
```

Sets Cannot Be Indexed

```python
# Topic: No Indexing in Sets
# Sets have no order, so we cannot use positions like [0]

s = {10, 20, 30}

try:
    print(s[0])
except TypeError as error:
    print("Error:", error)
```

Adding Items to a Set

```python
# Topic: add()
# add() puts one new item in the set

skills = {"Python", "SQL"}
skills.add("Pandas")
print(skills)

# Adding an item that already exists changes nothing
skills.add("Python")
print(skills)
```

Updating a Set

```python
# Topic: update()
# update() adds many items at once from a list, tuple or another set

skills = {"Python", "SQL"}
skills.update(["NumPy", "Matplotlib", "SQL"])
print(skills)
```

Removing Items from a Set

```python
# Topic: Removing Items
# discard() removes an item and gives NO error if it is missing
# remove() removes an item but gives an error if it is missing
# pop() removes a random item
# clear() empties the set

tools = {"VSCode", "PyCharm", "Colab", "Jupyter"}

tools.discard("Jupyter")
tools.discard("Notepad")      # not in the set, but no error
print("After discard:", tools)

tools.remove("PyCharm")
print("After remove:", tools)

try:
    tools.remove("Notepad")
except KeyError as error:
    print("remove() error for missing item:", error)

removed = tools.pop()
print("Popped:", removed, "| Set now:", tools)

tools.clear()
print("After clear:", tools)
```

Built-in Functions with Set

```python
# Topic: Built-in Functions with Sets

scores = {45, 78, 62, 90, 0}

print("all() - are all values non-zero?", all(scores))
print("any() - is at least one value non-zero?", any(scores))
print("len():", len(scores))
print("max():", max(scores))
print("min():", min(scores))
print("sum():", sum(scores))
print("sorted() gives a list:", sorted(scores))

for i, value in enumerate(sorted(scores)):
    print("enumerate:", i, value)
```

Iterating Over a Set

```python
# Topic: Looping Through a Set
# A for loop visits each item, but the order is not guaranteed

animals = {"lion", "tiger", "deer"}

for animal in animals:
    print(animal)
```

Finding Number of Set Elements

```python
# Topic: len()
# len() tells how many items the set has

letters = set("programming")   # duplicate letters are removed
print(letters)
print("Unique letters:", len(letters))
```

Set Operations: Union

```python
# Topic: Union
# Union combines all items from both sets (no duplicates)

class_a = {"Ali", "Sara", "Zain"}
class_b = {"Zain", "Hina", "Omar"}

print("Using | :", class_a | class_b)
print("Using union():", class_a.union(class_b))
```

Set Operations: Intersection

```python
# Topic: Intersection
# Intersection keeps only the items found in BOTH sets

class_a = {"Ali", "Sara", "Zain"}
class_b = {"Zain", "Hina", "Sara"}

print("Using & :", class_a & class_b)
print("Using intersection():", class_a.intersection(class_b))
```

Set Operations: Difference

```python
# Topic: Difference
# Difference keeps items that are in the first set but NOT in the second

class_a = {"Ali", "Sara", "Zain"}
class_b = {"Zain", "Hina"}

print("Using - :", class_a - class_b)
print("Using difference():", class_a.difference(class_b))
```

Set Operations: Symmetric Difference

```python
# Topic: Symmetric Difference
# Keeps items that are in only ONE of the sets, not in both

class_a = {"Ali", "Sara", "Zain"}
class_b = {"Zain", "Hina"}

print("Using ^ :", class_a ^ class_b)
print("Using symmetric_difference():", class_a.symmetric_difference(class_b))
```

Check if Two Sets are Equal

```python
# Topic: Comparing Sets
# == checks if both sets have exactly the same items (order does not matter)

x = {1, 2, 3}
y = {3, 2, 1}
z = {1, 2}

print("x == y:", x == y)
print("x == z:", x == z)
```

Other Set Methods

```python
# Topic: Other Useful Set Methods

a = {1, 2, 3, 4}
b = {3, 4, 5}

# copy() makes a separate copy of the set
c = a.copy()
print("Copy of a:", c)

# isdisjoint() is True if the sets have nothing in common
print("Are a and {7, 8} disjoint?", a.isdisjoint({7, 8}))

# issubset() and issuperset()
print("Is {1, 2} a subset of a?", {1, 2}.issubset(a))
print("Is a a superset of {1, 2}?", a.issuperset({1, 2}))

# The _update methods change the set itself
d = a.copy()
d.difference_update(b)
print("difference_update:", d)

e = a.copy()
e.intersection_update(b)
print("intersection_update:", e)

f = a.copy()
f.symmetric_difference_update(b)
print("symmetric_difference_update:", f)
```

Frozenset

```python
# Topic: Frozenset
# A frozenset is a set that cannot be changed after it is created

fixed = frozenset(["red", "green", "blue"])
print(fixed)

try:
    fixed.add("yellow")
except AttributeError as error:
    print("Error:", error)
```
