```python
# Task 13: Program to show how a nested if statement works
# A nested if means an if statement placed inside another if statement

# Store the value 10 in variable i
i = 10

# Outer if: check if i is equal to 10
if (i == 10):

    # First inner if: check if i is smaller than 15
    if (i < 15):
        print("i is smaller than 15")

    # Second inner (nested) if: this is only checked
    # because the outer if condition above is True
    if (i < 12):
        print("i is smaller than 12 too")
    else:
        # This runs only if i is not smaller than 12
        print("i is greater than 15")
```
