```python
# Task 7: Program to convert temperature from Celsius to Fahrenheit

# Store the temperature in Celsius
# Change this value to get a different result
celsius = 37.5

# Formula to convert Celsius into Fahrenheit: F = (C x 1.8) + 32
fahrenheit = (celsius * 1.8) + 32

# Display the result
# %0.1f shows the number with 1 digit after the decimal point
print('%0.1f degree Celsius is equal to %0.1f degree Fahrenheit' % (celsius, fahrenheit))
```
