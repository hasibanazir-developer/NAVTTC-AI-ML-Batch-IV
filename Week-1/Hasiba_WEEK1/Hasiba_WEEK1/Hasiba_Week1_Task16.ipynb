Basic While Loop

```python
# Topic: Basic While Loop
# A while loop keeps running its code again and again as long as the condition is True.
# When the condition becomes False, the loop stops.

# Start counting from 1
step = 1

# Keep printing until step goes past 5
while step <= 5:
    print("Step number:", step)
    # Increase step by 1, otherwise the loop would never end
    step += 1

print("Loop is over")
```

While Loop to Add Numbers

```python
# Topic: While Loop Example - Adding Numbers
# Here we use a while loop to add all numbers from 1 to 20

n = 1        # the number we are currently adding
total = 0    # this will hold the running total

while n <= 20:
    total += n   # add the current number to the total
    n += 1       # move to the next number

print("Sum of numbers from 1 to 20 is", total)
```

While Loop with a List

```python
# Topic: While Loop with a List
# A while loop can also go through a list until the list becomes empty

colors = ["red", "green", "blue"]

# An empty list counts as False, so the loop stops when nothing is left
while colors:
    # pop() takes out the last item and gives it back to us
    print("Removed color:", colors.pop())

print("Colors left in the list:", colors)
```

Single Statement While Loop

```python
# Topic: Single Statement While Loop
# If the loop body has only one line, it can be written on the same line as while

count = 3
while count > 0: count -= 1; print("Count is now", count)
```

Infinite While Loop

```python
# Topic: Infinite While Loop
# If the condition of a while loop never becomes False, the loop runs forever.
# This is called an infinite loop. "while True" is a common way to write one.
# To keep this example safe, we use break to stop it after a few rounds.

rounds = 0

while True:
    rounds += 1
    print("This loop has run", rounds, "time(s)")

    # Without this check, the loop would never stop
    if rounds == 4:
        print("Stopping the infinite loop with break")
        break
```

While Loop with Continue Statement

```python
# Topic: While Loop with Continue Statement
# continue skips the remaining code for this round and jumps to the next round

num = 0

while num < 10:
    num += 1   # increase first, so the loop does not get stuck

    # Skip all numbers that are multiples of 3
    if num % 3 == 0:
        continue

    print(num)
```

While Loop with Break Statement

```python
# Topic: While Loop with Break Statement
# break stops the loop right away, even if the condition is still True

balance = 1000   # money in the account
withdraw = 300   # money taken out each time

while balance > 0:
    # Stop if there is not enough money for the next withdrawal
    if balance < withdraw:
        print("Not enough balance for another withdrawal")
        break
    balance -= withdraw
    print("Withdrew", withdraw, "| Remaining balance:", balance)
```

While Loop with Pass Statement

```python
# Topic: While Loop with Pass Statement
# pass does nothing. It is useful when Python needs a line of code
# but we do not want anything to happen there yet.

word = "loop"
i = 0

while i < len(word):
    if word[i] == "o":
        pass   # we do nothing special for the letter 'o' right now
    else:
        print("Letter:", word[i])
    i += 1
```

While Loop with Else

```python
# Topic: While Loop with Else
# The else part runs only when the loop ends normally, without hitting break

# Example 1: the loop finishes normally, so the else part runs
x = 1
while x <= 3:
    print("x =", x)
    x += 1
else:
    print("Loop ended normally, so else runs")

# Example 2: the loop is stopped by break, so the else part is skipped
y = 1
while y <= 5:
    if y == 3:
        print("Breaking at y =", y)
        break
    print("y =", y)
    y += 1
else:
    print("This will not be printed")
```

Practice: Guessing a Secret Number

```python
# Topic: Practice - Searching with a While Loop
# The loop checks numbers one by one until it finds the secret number

secret = 7
guess = 1

while guess != secret:
    print(guess, "is not the secret number")
    guess += 1

print("Found it! The secret number is", guess)
```
