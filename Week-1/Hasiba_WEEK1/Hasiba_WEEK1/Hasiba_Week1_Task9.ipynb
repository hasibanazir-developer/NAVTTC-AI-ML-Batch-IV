```python
# Task 9: Program to solve the quadratic equation ax**2 + bx + c = 0

# Import the cmath module so we can also handle complex (imaginary) answers
import cmath

# Store the values of the coefficients a, b and c
a = 1
b = 5
c = 6

# Calculate the discriminant using the formula: d = b^2 - 4ac
# The discriminant tells us what type of solutions the equation has
d = (b ** 2) - (4 * a * c)

# Find the two solutions using the quadratic formula: (-b ± square root of d) / 2a
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

# Display both solutions
print('The solutions are {0} and {1}'.format(sol1, sol2))
```
