```python
# Task 11: Program to show how a simple if statement works

# Store the value 10 in variable i
i = 10

# Check if i is greater than 15
# Since 10 is not greater than 15, the condition is False,
# so the code inside the if block will NOT run
if (i > 15):
    print("10 is less than 15")

# This line is outside the if block, so it always runs
print("I am Not in if")
```
