```python
# Task 14: Program to show how an if-elif-else ladder works
# Python checks each condition from top to bottom
# and runs only the first block whose condition is True

# Store the value 20 in variable i
i = 20

# Check if i is 10
if (i == 10):
    print("i is 10")

# If not, check if i is 15
elif (i == 15):
    print("i is 15")

# If not, check if i is 20 (this condition is True, so this block runs)
elif (i == 20):
    print("i is 20")

# If none of the above conditions are True, this block runs
else:
    print("i is not present")
```
