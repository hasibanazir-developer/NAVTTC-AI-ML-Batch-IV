```python
# Task 6: Program to find the square root of real or complex numbers

# Import the cmath module, which handles complex numbers
# A complex number has a real part and an imaginary part (written with j)
import cmath

# Store a complex number (1 is the real part, 2j is the imaginary part)
num = 1+2j

# Uncomment the line below to take the number from the user
# eval() converts the input text into a number (including complex numbers)
# num = eval(input('Enter a number: '))

# Find the square root using the sqrt() function from cmath
num_sqrt = cmath.sqrt(num)

# Display the result
# .real gives the real part and .imag gives the imaginary part
# :0.3f shows each part with 3 digits after the decimal point
print('The square root of {0} is {1:0.3f}+{2:0.3f}j'.format(num, num_sqrt.real, num_sqrt.imag))
```
