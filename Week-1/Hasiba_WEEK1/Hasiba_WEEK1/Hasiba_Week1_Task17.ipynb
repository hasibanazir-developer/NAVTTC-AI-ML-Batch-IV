What is the Break Statement

```python
# Topic: Break Statement
# break is used to stop a loop early.
# As soon as Python reaches break, it leaves the loop and moves on.

for letter in "hello world":
    # Stop as soon as we reach the space
    if letter == " ":
        break
    print(letter)

print("We left the loop at the space")
```

Break Statement with For Loop

```python
# Topic: Break with For Loop
# Here the loop should run from 1 to 10, but break stops it at 6

for num in range(1, 11):
    if num == 6:
        print("Reached 6, stopping the loop")
        break
    print(num)
```

Break with For Loop: Searching a List

```python
# Topic: Break with For Loop - Searching
# break is very handy when we are searching for something.
# Once we find it, there is no need to keep checking.

students = ["Ayesha", "Bilal", "Haseeba", "Usman", "Zara"]
target = "Haseeba"

for position, name in enumerate(students):
    if name == target:
        print(target, "found at position", position)
        break
    print("Checked", name)
```

Break Statement with While Loop

```python
# Topic: Break with While Loop
# A countdown that stops when it reaches zero

timer = 5

while True:
    print("Time left:", timer)
    timer -= 1

    # Leave the loop when the countdown is finished
    if timer == 0:
        print("Time is up!")
        break
```

Break with While Loop: Sum Limit

```python
# Topic: Break with While Loop - Stop at a Limit
# Keep adding even numbers, but stop once the total goes above 50

total = 0
number = 2

while number <= 100:
    total += number
    if total > 50:
        print("Total went above 50 after adding", number)
        break
    number += 2

print("Final total:", total)
```

Break in Nested Loops

```python
# Topic: Break in Nested Loops
# break only stops the loop it is written in (the inner loop).
# The outer loop keeps running as normal.

for i in range(1, 4):
    for j in range(1, 4):
        if j == 2:
            break   # this only stops the inner loop
        print("i =", i, "j =", j)
    print("Inner loop stopped, outer loop continues with i =", i)
```

Leaving Both Loops Using a Flag

```python
# Topic: Breaking Out of Both Loops with a Flag
# To stop both loops at once, we can use a True/False variable called a flag

grid = [[1, 3, 5], [7, 8, 9], [11, 13, 15]]
found = False   # this flag tells us if an even number was found

for row in grid:
    for value in row:
        if value % 2 == 0:
            print("First even number found:", value)
            found = True
            break      # stop the inner loop
    if found:
        break          # stop the outer loop as well

print("Search finished")
```
