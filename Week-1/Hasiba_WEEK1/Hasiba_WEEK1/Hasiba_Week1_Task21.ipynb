Creating a List

```python
# Topic: Creating a List
# A list stores many items in one variable, inside square brackets

marks = [85, 90, 78, 92]
mixed = ["Haseeba", 21, 5.4, True]   # a list can hold different types
empty = []                           # an empty list

# list() can also create a list, for example from a string
letters = list("AI")

print(marks)
print(mixed)
print(empty)
print(letters)
```

Accessing List Items

```python
# Topic: Accessing List Items
# Each item has a position called an index. Indexing starts from 0.

languages = ["Python", "Java", "C++", "JavaScript"]

print("First item:", languages[0])
print("Third item:", languages[2])
```

Negative Indexing

```python
# Topic: Negative Indexing
# Negative numbers count from the end. -1 is the last item.

languages = ["Python", "Java", "C++", "JavaScript"]

print("Last item:", languages[-1])
print("Second last item:", languages[-2])
```

Slicing a List

```python
# Topic: Slicing a List
# list[start:stop] gives items from start up to stop-1

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(days[1:4])   # Tue to Thu
print(days[:3])    # first three days
print(days[4:])    # from Fri to the end
print(days[::2])   # every second day
```

Adding Items to a List

```python
# Topic: Adding Items
# append() adds one item at the end
# insert() adds an item at a chosen position
# extend() adds many items from another list

cart = ["bread", "milk"]

cart.append("eggs")
print("After append:", cart)

cart.insert(1, "butter")
print("After insert:", cart)

cart.extend(["rice", "sugar"])
print("After extend:", cart)
```

Updating List Items

```python
# Topic: Changing (Updating) Items
# We can replace an item by assigning a new value to its index

cart = ["bread", "milk", "eggs"]
cart[1] = "yogurt"
print("Updated list:", cart)
```

Removing Items from a List

```python
# Topic: Removing Items
# remove() deletes an item by its value
# pop() deletes an item by its index (the last one if no index is given)
# del deletes an item or a range of items
# clear() empties the whole list

pets = ["cat", "dog", "parrot", "fish", "rabbit"]

pets.remove("parrot")
print("After remove:", pets)

last = pets.pop()
print("Popped item:", last, "| List now:", pets)

del pets[0]
print("After del:", pets)

pets.clear()
print("After clear:", pets)
```

Copying a List

```python
# Topic: Copying a List
# Using = does NOT make a real copy. Both names point to the same list.
# copy() creates a separate new list.

original = [1, 2, 3]

same = original          # not a real copy
same.append(4)
print("Original after changing 'same':", original)

real_copy = original.copy()
real_copy.append(5)
print("Original:", original)
print("Real copy:", real_copy)
```

List Methods: index(), count(), sort(), reverse()

```python
# Topic: More List Methods

nums = [4, 2, 9, 2, 7, 2]

print("Index of 9:", nums.index(9))       # position of the first 9
print("How many times 2 appears:", nums.count(2))

nums.sort()
print("Sorted (ascending):", nums)

nums.sort(reverse=True)
print("Sorted (descending):", nums)

nums.reverse()
print("Reversed:", nums)
```

The len() Function

```python
# Topic: len()
# len() tells us how many items are in a list

books = ["Book A", "Book B", "Book C"]
print("Number of books:", len(books))
```

List Membership Test

```python
# Topic: Membership Test
# 'in' checks if a value is in the list, 'not in' checks if it is missing

fruits = ["apple", "banana", "mango"]

print("Is mango in the list?", "mango" in fruits)
print("Is grapes not in the list?", "grapes" not in fruits)
```

Iterating Through a List

```python
# Topic: Iterating Through a List
# A for loop visits each item of the list one by one

cities = ["Lahore", "Faisalabad", "Peshawar"]

for city in cities:
    print("I want to visit", city)
```
