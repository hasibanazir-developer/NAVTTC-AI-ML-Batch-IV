Creating a String

```python
# Topic: Creating a String
# A string is text written inside single or double quotes

greeting = "Assalam o Alaikum"
course = 'Artificial Intelligence'

print(greeting)
print(course)
```

Multiline Strings

```python
# Topic: Multiline Strings
# Triple quotes let us write a string over many lines

note = """Week 1 topics:
Python basics
Loops and conditions
Data structures"""

print(note)
```

Accessing String Characters: Indexing

```python
# Topic: Indexing
# Each character has a position. The first one is at index 0,
# and -1 is the last character.

word = "Machine"

print("First character:", word[0])
print("Fourth character:", word[3])
print("Last character:", word[-1])
print("Second last character:", word[-2])
```

Accessing String Characters: Slicing

```python
# Topic: Slicing
# string[start:stop] gives characters from start up to stop-1

text = "DeepLearning"

print(text[0:4])    # Deep
print(text[4:])     # Learning
print(text[:4])     # Deep
print(text[::-1])   # the whole string reversed
```

Strings are Immutable

```python
# Topic: Strings Cannot Be Changed
# We cannot change a single character of a string.
# To change it, we make a new string.

name = "hasiba"

try:
    name[0] = "H"
except TypeError as error:
    print("Error:", error)

# Create a new string instead
name = "H" + name[1:]
print("New string:", name)
```

String Operations: Joining and Repeating

```python
# Topic: Joining (Concatenation) and Repeating Strings
# + joins two strings together, * repeats a string

first = "Data"
second = "Science"

print(first + " " + second)
print("Go! " * 3)
```

Comparing Strings

```python
# Topic: Comparing Strings
# == checks if two strings are exactly the same

a = "python"
b = "Python"

print("Are they equal?", a == b)
print("Equal after lowercase?", a == b.lower())
```

Python String Methods

```python
# Topic: String Methods
# Methods return a NEW string; the original string stays the same

sentence = "  learning python is fun  "

print(sentence.upper())             # all capital letters
print(sentence.lower())             # all small letters
print(sentence.strip())             # remove spaces from both sides
print(sentence.replace("fun", "great"))
print(sentence.split())             # break into a list of words
print("-".join(["AI", "ML", "DL"])) # join a list into one string
print(sentence.find("python"))      # position of the word
print(sentence.count("n"))          # how many times 'n' appears
print("hello world".title())        # first letter of each word capital
print("2026".isdigit())             # True if all characters are digits
print("report.pdf".endswith(".pdf"))
```

String Membership Test

```python
# Topic: Membership Test
# 'in' and 'not in' check if a word or letter is inside a string

message = "Welcome to the AI course"

print("Is 'AI' in the message?", "AI" in message)
print("Is 'Java' not in the message?", "Java" not in message)
```

Iterate Through a String

```python
# Topic: Iterating Through a String
# A for loop gives us one character at a time

for ch in "Code":
    print(ch)
```

String Length

```python
# Topic: String Length
# len() counts the number of characters, including spaces

title = "Neural Networks"
print("Number of characters:", len(title))
```

Escape Sequences

```python
# Topic: Escape Sequences
# A backslash (\) lets us add special characters inside a string

print("She said, \"Python is easy!\"")   # double quotes inside a string
print('It\'s a nice day')                 # single quote inside a string
print("Line one\nLine two")               # \n means new line
print("Name:\tHaseeba")                   # \t means a tab space
print("C:\\Users\\Documents")              # \\ prints one backslash
```

String Formatting with f-Strings

```python
# Topic: f-Strings
# Put f before the quotes and write variables inside curly brackets {}

name = "Haseeba"
week = 1
tasks_done = 23

print(f"{name} has completed {tasks_done} tasks in week {week}.")
print(f"Tasks left out of 25: {25 - tasks_done}")

# format() is an older way to do the same thing
print("{} is in week {}".format(name, week))
```
