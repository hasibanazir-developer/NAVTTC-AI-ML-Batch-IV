```python
# Task 8: Program to find the area of a triangle using Heron's formula

# Store the length of the three sides of the triangle
a = 5
b = 6
c = 7

# Uncomment the lines below to take the sides from the user
# a = float(input('Enter first side: '))
# b = float(input('Enter second side: '))
# c = float(input('Enter third side: '))

# Calculate the semi-perimeter (half of the total length of all sides)
s = (a + b + c) / 2

# Calculate the area using Heron's formula: area = square root of s(s-a)(s-b)(s-c)
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

# Display the area with 2 digits after the decimal point
print('The area of the triangle is %0.2f' % area)
```
