Checking the Python Version

```python
# Topic: Checking Python
# Before installing anything, we check which version of Python we have.
# Anaconda, PyTorch and TensorFlow all run on top of Python.

import sys
import platform

print("Python version:", platform.python_version())
print("Full details:", sys.version)
```

Download and Install Anaconda3

Anaconda is a free package that installs Python together with many data science libraries (NumPy, Pandas, Jupyter and more). It is installed on the computer, not on Colab.

**Steps:**
1. Open https://www.anaconda.com/download and download the Windows installer.
2. Right-click the installer and choose **Run as administrator**.
3. Click **Next**, then **I Agree**, then select **Just Me (recommended)**.
4. Keep the install location simple, for example `C:\anaconda3`, and click **Install**.
5. When it finishes, open **Anaconda Navigator** from the Start menu.

```python
# Topic: What Anaconda Gives Us
# Anaconda comes with many ready-made libraries.
# Colab also has most of them already, so we can check a few here.

import numpy, pandas, matplotlib, sklearn

print("NumPy version:", numpy.__version__)
print("Pandas version:", pandas.__version__)
print("Matplotlib version:", matplotlib.__version__)
print("Scikit-learn version:", sklearn.__version__)
```

Install PyTorch

```python
# Topic: Installing PyTorch
# PyTorch is a deep learning library made by Meta.
# On a computer we install it with: pip install torch
# In Colab it is usually already installed, so this just confirms it.

!pip install torch
```

```python
# Topic: Checking PyTorch
# We import PyTorch, print its version and do a small test calculation

import torch

print("PyTorch version:", torch.__version__)

# A tensor is like a NumPy array that PyTorch uses
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])
print("Tensor a:", a)
print("Tensor b:", b)
print("a + b =", a + b)

# Check if a GPU is available (in Colab: Runtime > Change runtime type > GPU)
print("GPU available for PyTorch:", torch.cuda.is_available())
```

Install TensorFlow 2.0

```python
# Topic: Installing TensorFlow
# TensorFlow is a deep learning library made by Google.
# On a computer we install it with: pip install tensorflow
# The latest version is from the 2.x series, which includes TensorFlow 2.0 features.

!pip install tensorflow
```

```python
# Topic: Checking TensorFlow
# We import TensorFlow, print its version and do a small test calculation

import tensorflow as tf

print("TensorFlow version:", tf.__version__)
print("Is it TensorFlow 2.x?", tf.__version__.startswith("2"))

x = tf.constant([[1, 2], [3, 4]])
y = tf.constant([[5, 6], [7, 8]])
print("Matrix multiplication result:")
print(tf.matmul(x, y).numpy())

print("GPUs found by TensorFlow:", tf.config.list_physical_devices("GPU"))
```
